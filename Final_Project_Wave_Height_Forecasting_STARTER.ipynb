## The dataset

Real, publicly published data from **NOAA's National Data Buoy Center (NDBC)**, station **46026** — "San Francisco, 18 NM West of San Francisco, CA" (37.750°N, 122.838°W), a moored ocean buoy in 53 m of water. This is genuine oceanographic sensor data, not a curated teaching dataset — it has the gaps, noise, and quirks real sensor data always has (already partially cleaned for you, see below).

| Column | Meaning | Units |
|---|---|---|
| `year`, `month`, `day`, `hour`, `minute` | Timestamp of the reading (UTC) | — |
| `WDIR` | Wind direction | degrees true |
| `WSPD` | Wind speed | m/s |
| `GST` | Wind gust speed | m/s |
| `PRES` | Sea-level atmospheric pressure | hPa |
| `WTMP` | Sea water temperature | °C |
| `WVHT` | **Significant wave height — your prediction target** | meters |

Two other real wave measurements this station records — dominant wave period and wave direction — are **deliberately excluded** from the features you're given. They describe the same sea state as `WVHT` itself and would make the task trivially easy in a way that wouldn't reflect a genuine forecasting problem (the same leakage principle from `NB03` §5: a feature that's really just another view of the target doesn't teach a model anything useful). You are working with wind, pressure, and water temperature alone — the variables that are physically upstream of wave formation, not measurements of the waves themselves.

## The task

Using `train.csv` (real readings from **2019–2023**), build a model that predicts `WVHT` for the readings in `test_features.csv` (real readings from **2024** — a genuine future period the training data never saw). This is a **temporal holdout**, not a random split — exactly the kind of realistic evaluation `NB10` used for its own forecasting task.

## Evaluation

Your predictions are scored by **Root Mean Squared Error (RMSE)** between your predicted `WVHT` and the true (held-out, private) values — lower is better, `NB03`'s regression metric from Part 3. Submit a CSV with exactly two columns, `row_id` and `WVHT_pred`, one row per row in `test_features.csv`, in any order, using the `row_id` column to match them up.
