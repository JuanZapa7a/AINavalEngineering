The **Ship Fuel Consumption Dataset** is a dataset which provides a detailed overview of fuel consumption and CO2 emissions for various ship types operating in Nigerian waterways. It includes data on ship types, routes, engine efficiency, fuel consumption, Month and emissions, making it suitable for environmental impact studies, maritime operations optimization, and predictive modeling.

## **Overview**  
  
This project analyzes fuel consumption and CO2 emissions of various ship types operating in Nigerian waterways. By exploring the fuel efficiency and environmental impact of these vessels, we aim to provide actionable insights for optimizing maritime operations and reducing emissions.

## **Dataset Structure**  

The dataset used in this project contains the following columns:

  - Ship Type: Categorizes ships into four main types: Fishing Trawler, Oil Service Boat, Surfer Boat, and Tanker Ship.
  - Fuel Consumption (Liters): The total fuel consumed by each ship type during operations.
  - CO2 Emission (Kg): The amount of carbon dioxide emitted based on fuel consumption.
  - Other Variables: Supporting data used for exploratory analysis.

This dataset was generated to simulate realistic maritime operations in Nigeria, taking into account common ship types, fuel usage patterns, and emissions.

## **Key Insights**  

1. Fuel Consumption by Ship Type:
  - Tanker Ships have the highest average fuel consumption, reflecting their larger size and cargo capacity.
  - Surfer Boats consume the least fuel, making them efficient for shorter, high-speed trips.
2. CO2 Emissions:
  - A strong positive correlation (r ≈ 0.997) exists between fuel consumption and CO2 emissions, confirming that higher fuel usage results in greater emissions.
3. Statistical Analysis:
  - ANOVA tests reveal significant differences in fuel consumption across ship types.
  - Tukey HSD analysis pinpoints which ship types differ in fuel consumption and emissions, providing actionable insights.

**Check Missing Values**

---

Missing values (`NaN`, `None`, or placeholders like `-999`) in a dataset can significantly impact data quality, analysis, and model performance. Here’s why you should always check for them:  

1. **Avoid Biased Analysis**  
   - Missing data can skew statistics (e.g., `mean()`, `median()`).  
   - Example: If 50% of `fuel_consumption` values are missing, the average will be unreliable.  

2. **Prevent Model Errors**  
   - Most machine learning algorithms (e.g., scikit-learn) fail if missing values are present.  
   - Example: A linear regression predicting `CO2_emissions` will error out if `distance` has `NaN`s.  

3. **Improve Data Quality**  
   - Missing values may indicate data collection issues (e.g., sensor failure, human error).  
   - Example: Empty `weather_conditions` could mean faulty logging systems.  

4. **Choose the Right Handling Strategy**  
   Depending on the context, you might:  
   - **Drop rows/columns**: If missing values are rare (e.g., `df.dropna()`).  
   - **Impute**: Fill gaps with mean/median (numeric) or mode (categorical).  
     ```python
     df['distance'].fillna(df['distance'].mean(), inplace=True)
     ```  
   - **Flag missingness**: Create a binary column (e.g., `is_missing_fuel_type`).

**Identify outliers: IQR (Interquartile Range) method**

---

The Importance of Identifying Outliers

1. **Statistical Analysis Impact**:
   - Skew measures like mean and standard deviation
   - Example: A single extreme value in `fuel_consumption` can inflate the average

2. **Machine Learning Issues**:
   - Algorithms like linear regression are sensitive to extreme values
   - Can reduce predictive model accuracy

3. **Data Quality Control**:
   - May indicate data collection errors
   - Example: A `distance` value of 5000 when normal max is 500

4. **Valuable Pattern Discovery**:
   - Sometimes represent exceptional but meaningful cases
   - Example: A ship with abnormally high `engine_efficiency` might reveal best practices

`````python
# For no-normal (gauss) distributions
def count_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25) # First quartile
    Q3 = df[column].quantile(0.75) # Third quartile
    IQR = Q3 - Q1 # Interquartile range
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return len(outliers)

# Apply to all numeric columns
numeric_cols = df.select_dtypes(include=['number']).columns
outliers_count = {col: count_outliers_iqr(df, col) for col in numeric_cols}

print("Outliers Detected:(IQR method):")
for col, count in outliers_count.items():
    print(f"{col}: {count} outliers")
`````

---
**Transformations (Using log1p)**

---

It is necessary applies a **log transformation** to handle **skewed data**. `np.log1p(x)` calculates `log(1 + x)` - it's the natural logarithm of (1 + x). Because:

**1. Reduces skewness**
- Many real-world features (like emissions, consumption, distance) are **right-skewed** - they have a long tail of high values
- The log transformation compresses large values more than small ones, making the distribution more **normal/symmetric**

**2. Handles zeros safely**
- Regular `log(x)` fails when x = 0 (undefined)
- `log1p(x) = log(1 + x)` works perfectly when x = 0, giving you log(1) = 0

**3. Improves model performance**
- Many machine learning algorithms (especially linear models) work better with normally-distributed features
- It reduces the impact of outliers
- Makes relationships between variables more linear

### Example

```
Original values: [0, 10, 100, 1000]
After log1p:     [0, 2.4, 4.6, 6.9]
```

Notice how the huge gap between 100 and 1000 becomes much smaller (4.6 to 6.9), while the scale of smaller values is preserved.

### In our specific case

You're transforming:
- **CO2_emissions** - likely has some very high-polluting vehicles
- **fuel_consumption** - some vehicles consume much more than others
- **distance** - probably has some very long trips

This makes these features more suitable for modeling and prevents extreme values from dominating your model's predictions.

`````python
import numpy as np

# Apply log1p transformation to skewed features
df['CO2_emissions'] = np.log1p(df['CO2_emissions'])
df['fuel_consumption'] = np.log1p(df['fuel_consumption'])
df['distance'] = np.log1p(df['distance'])
`````

`````python
# For no-normal (gauss) distributions
def count_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25) # First quartile
    Q3 = df[column].quantile(0.75) # Third quartile
    IQR = Q3 - Q1 # Interquartile range
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return len(outliers)

# Apply to all numeric columns
numeric_cols = df.select_dtypes(include=['number']).columns
outliers_count = {col: count_outliers_iqr(df, col) for col in numeric_cols}

print("Outliers Detected:(IQR method):")
for col, count in outliers_count.items():
    print(f"{col}: {count} outliers")
`````

---
**Scale numerical values and Transform categorical values (OneHot)**

---

**OneHotEncoder** converts categorical variables (text labels) into a numerical format that machine learning models can understand.

## **The Problem**

Machine learning models need **numbers**, not text. But you can't just convert categories to numbers like:
```
Red = 1, Blue = 2, Green = 3  ❌ WRONG!
```

This implies **Blue > Red** and **Green > Blue**, which creates a false ordering/relationship.

## **The Solution: One-Hot Encoding**

Creates a **separate binary column** for each category:

### **Example:**

**Original data:**
```
| Car Type |
|----------|
| Sedan    |
| SUV      |
| Sedan    |
| Truck    |
```

**After OneHotEncoding:**
```
| Car_Type_Sedan | Car_Type_SUV | Car_Type_Truck |
|----------------|--------------|----------------|
|       1        |      0       |       0        |
|       0        |      1       |       0        |
|       1        |      0       |       0        |
|       0        |      0       |       1        |
```

Each row has a **1** in the column that matches its category, and **0** everywhere else.

### **What `handle_unknown='ignore'` does**

If your test data has a category that **wasn't in the training data**:
- **'ignore'**: Creates a row of all zeros (doesn't crash)
- Without it: Would throw an error

#### **Example:**
```
Training: [Sedan, SUV, Truck]
Test data includes: Van  ← unknown category

With handle_unknown='ignore':
Van → [0, 0, 0]  ✅ Safe, continues working
```

### **In Our Case**

Our `ColumnTransformer`:
- Applies **StandardScaler** to numeric columns (normalizes numbers)
- Applies **OneHotEncoder** to categorical columns (converts text to binary columns)

`````python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numeric_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_cols)
    ])
`````

---
**Test Options and Evaluation Metric**

---

This code sets up **cross-validation** and **evaluation metrics** for our machine learning model.

### KFold Cross-Validation

Splits your data into 10 parts (folds) to test our model more reliably.

**How it works:**
1. Divides data into 10 equal parts
2. Trains on 9 parts, tests on 1 part
3. Repeats 10 times, using each part as test data once
4. Averages the results

**Parameters:**
- **`n_splits=10`**: Creates 10 folds (10-fold cross-validation)
- **`shuffle=True`**: Randomly shuffles data before splitting (avoids bias if data is ordered)
- **`random_state=42`**: Makes results reproducible (same random shuffle every time)

**Why use it?** More reliable than a single train/test split - you test on ALL your data eventually.

### Evaluation Metrics

We are calculating **3 metrics** to evaluate our model performance:

**1. neg_mse (Negative Mean Squared Error)**
- **Formula:** `-average((predicted - actual)²)`
- **Range:** -∞ to 0 (closer to 0 is better)
- **Why negative?** sklearn's cross_val_score maximizes scores, but we want to minimize error, so it uses negative values
- **Use:** Penalizes large errors heavily (because of squaring)

**2. r2 (R-squared / Coefficient of Determination)**
- **Range:** 0 to 1 (sometimes negative for very bad models)
- **Meaning:** Percentage of variance in target variable explained by your model
- **Interpretation:**
  - **r² = 0.85** → Model explains 85% of the variance
  - **r² = 1.0** → Perfect predictions
  - **r² = 0.0** → Model is no better than predicting the mean
- **Use:** Easy to interpret overall model quality

**3. rmse (Root Mean Squared Error)**
- **Formula:** `√(average((predicted - actual)²))`
- **Units:** Same as your target variable (e.g., if predicting CO2 in grams, RMSE is in grams)
- **Use:** Shows average prediction error in interpretable units
- **Example:** RMSE = 15 means predictions are off by about 15 units on average

**Why all three?**

- **MSE/RMSE**: Show prediction accuracy in absolute terms
- **R²**: Shows how well your model fits the data relative to a baseline
- Together they give a complete picture of model performance

This ensures your model performs well on different subsets of data, not just one lucky train/test split!

`````python
from sklearn.model_selection import KFold, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.metrics import make_scorer, mean_squared_error, r2_score

# Standardize features and setup evaluation metrics
kfold = KFold(n_splits=10, shuffle=True, random_state=42)

# Define the metrics we want to calculate
scoring = {
    'neg_mse': 'neg_mean_squared_error',
    'r2': 'r2',
    'rmse': make_scorer(lambda y_true, y_pred: np.sqrt(mean_squared_error(y_true, y_pred)))
}
`````

### Build Models

Let’s create a baseline of performance on this problem and spot-check a number of different
algorithms. We will select a suite of different algorithms capable of working on this regression
problem. The six algorithms selected include:
- Linear Algorithms: Linear Regression (LR), Ridge Regression (RR), Lasso Regression.
- Nonlinear Algorithms: Decision Tree (DT), Random Forest (RF), Gradient Boosting (GB), Support Vector Regression (SVR)).

---
**Spot-Check Algorithms**

---

`````python
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR

# Baseline models
models = {
    'Linear Regression': LinearRegression(),
    'Ridge Regression': Ridge(alpha=1.0),
    'Lasso Regression': Lasso(alpha=0.1),
    #'Decision Tree': DecisionTreeRegressor(max_depth=5),
    'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42),
    'Gradient Boosting': GradientBoostingRegressor(n_estimators=100, random_state=42),
    'Support Vector': SVR(kernel='rbf', C=100)
}
`````

---
**Evaluate each model in turn**

---

`````python
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline

results = {}

# Iterate over the dictionary of models
for name, model in models.items():
    # Create a pipeline that includes preprocessing and the model itself
    pipeline = Pipeline([
        ('preprocessor', preprocessor),
        ('model', model)
    ])

    # Dictionary to store results for the current model
    model_results = {}

    try:
        # Evaluate the model using each scoring metric
        for metric_name, metric in scoring.items():
            cv_scores = cross_val_score(
                pipeline,
                X_train,
                y_train,
                cv=kfold,
                scoring=metric,
                error_score='raise'  # Raise an error immediately if any occurs
            )

            # Store results in a readable format
            if metric_name == 'neg_mse':
                model_results['MSE'] = -cv_scores.mean()  # Convert negative MSE to positive
                model_results['Std MSE'] = cv_scores.std()
            elif metric_name == 'r2':
                model_results['R2'] = cv_scores.mean()
                model_results['Std R2'] = cv_scores.std()
            elif metric_name == 'rmse':
                model_results['RMSE'] = cv_scores.mean()
                model_results['Std RMSE'] = cv_scores.std()

        # Save results for the current model
        results[name] = model_results

    except Exception as e:
        # Handle and store any errors that occur during evaluation
        print(f"Error with model {name}: {str(e)}")
        results[name] = {'Error': str(e)}
`````
