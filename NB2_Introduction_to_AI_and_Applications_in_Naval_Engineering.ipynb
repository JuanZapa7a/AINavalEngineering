## **Basic Concepts of AI**  
Artificial Intelligence refers to the simulation of human intelligence in machines that are programmed to think, learn, and make decisions. AI encompasses several subfields, each with unique applications and methodologies:  

1. **[Machine Learning (ML)](https://en.wikipedia.org/wiki/Machine_learning)**:  
   - A subset of AI that focuses on enabling machines to learn from data without being explicitly programmed. ML algorithms identify patterns and make predictions based on historical data.  
   - Example: [Predicting equipment failures based on sensor data](https://spd.tech/machine-learning/predictive-maintenance/).  

2. **[Artificial Neural Networks (ANNs)](https://en.wikipedia.org/wiki/Artificial_neural_network)**:  
   - Inspired by the human brain, ANNs are computational models used for tasks like image recognition, natural language processing, and decision-making.  
   - Example: [Autonomous navigation systems for ships](https://promwad.com/news/navigation-systems-autonomous-ships).  

3. **[Natural Language Processing (NLP)](https://en.wikipedia.org/wiki/Natural_language_processing)**:  
   - A branch of AI that enables machines to understand, interpret, and generate human language.  
   - Example: [Voice-controlled systems for ship operations](https://www.asti-usa.com/progexs/military/naval_comms.html).

## **Types of AI: Weak vs Strong AI**

Artificial Intelligence (AI) is broadly categorized into two main types: [Weak AI and Strong AI](https://www.geeksforgeeks.org/artificial-intelligence/what-is-the-difference-between-a-strong-ai-and-a-weak-ai/), representing distinct capabilities and levels of intelligence.


1. **Weak AI** (Narrow AI)

- **Definition**: AI systems designed and trained to perform *specific tasks*.
- **Capabilities**:
  - Focused on one domain.
  - Cannot perform tasks outside their training.
  - Lacks consciousness and understanding.
- **Examples**:
  - Siri, Alexa, ChatGPT
  - Spam filters
  - Netflix/YouTube recommendation engines
  - Image recognition systems
- **Key Characteristics**:
  - Task-specific
  - Efficient but limited
  - No real intelligence or awareness

Weak AI, also known as Narrow AI or specialized AI, is designed to perform specific, predefined tasks efficiently without possessing human-like general intelligence.

It operates within a limited context and cannot transfer learning to unrelated tasks.

Weak AI relies on predefined rules, algorithms, and data-driven learning to complete its functions.

It excels at specialized functions such as speech recognition, image processing, playing chess, or providing recommendations.

Common examples include virtual assistants like Siri and Alexa, recommendation systems used by Netflix and Amazon, chatbots for customer service, fraud detection systems, self-driving cars, and GPS navigation apps.

While highly effective within its narrow scope, Weak AI lacks consciousness, self-awareness, and the ability to understand or reason beyond its programmed parameters.

It simulates intelligence by processing data but does not possess true understanding.


2. **Strong AI** (General AI)

- **Definition**: A theoretical form of AI with *human-level intelligence*.
- **Capabilities**:
  - Can reason, plan, learn, and understand context.
  - Performs across multiple domains without retraining.
  - Possesses consciousness and self-awareness (in theory).
- **Examples**:
  - Not yet realized in practice.
  - Fictional: HAL 9000 (*2001: A Space Odyssey*), Data (*Star Trek*), Ava (*Ex Machina*)
- **Key Characteristics**:
  - General-purpose
  - Adaptive and flexible
  - Capable of abstract thought and decision-making

_**Summary Table**_

| Feature              | **Weak AI**                    | **Strong AI**                      |
|----------------------|--------------------------------|------------------------------------|
| Scope                | Narrow, task-specific          | Broad, general-purpose             |
| Consciousness        | No                             | Theoretical, Yes                   |
| Human-level thinking | No                             | Yes                                |
| Examples             | Siri, Google Translate         | HAL 9000, Data (*fictional*)       |
| Learning & Adapting  | Limited                        | Flexible and self-improving        |


**Key Insight**

> **Weak AI powers most of today’s applications** and excels in focused domains.  
> **Strong AI remains a future goal**, raising complex challenges in ethics, safety, and philosophy.

## **Applications of AI in Naval Engineering**  
AI is transforming naval engineering by enhancing efficiency, safety, and sustainability. Below are some key applications:  

1. **Automation**:  
   - AI enables the development of autonomous systems for ships, such as self-navigating vessels and automated control systems.  
   - Example: **[Autonomous Ships](https://www.marineinsight.com/tech/autonomous-ships-the-future-of-shipping/)** use AI to navigate, avoid obstacles, and optimize routes without human intervention.  

2. **Predictive Maintenance**:  
   - AI models analyze data from sensors and equipment to predict potential failures before they occur. This reduces downtime and maintenance costs.  
   - Example: Using **[Machine Learning](https://towardsdatascience.com/predictive-maintenance-using-machine-learning-2b8d6b02e5a4)** to monitor engine performance and detect anomalies in real-time.  

3. **Route Optimization**:  
   - AI algorithms optimize shipping routes by considering factors like weather conditions, fuel consumption, and delivery schedules.  
   - Example: **[AI-powered route planning](https://www.ibm.com/blogs/internet-of-things/iot-maritime-route-optimization/)** minimizes fuel usage and reduces environmental impact.  

4. **Data Analysis**:  
   - Ships generate vast amounts of data from sensors, navigation systems, and operational logs. AI processes this data to provide actionable insights.  
   - Example: Real-time monitoring of ship performance using **[Big Data Analytics](https://www.sciencedirect.com/topics/engineering/big-data-analytics)**.

### **Example: _Predicting Failures due to Engine Vibrations_**


```python
# ## Let's Visualize a Use Case

# %%
import matplotlib.pyplot as plt
import numpy as np

# Simulated engine vibration data
np.random.seed(0)
hours = np.arange(0, 24, 1)
vibration = 0.3 * np.sin(0.3 * hours) + np.random.normal(0, 0.1, size=hours.shape)

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, marker='o')
plt.title("Simulated Engine Vibration Over 24 Hours")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.grid(True)
plt.show()
```

If we label when vibration goes above a threshold, AI can learn to predict failures in advance.

```python
# %%
thresh = 0.25
alerts = vibration > thresh

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, label="Vibration")
plt.axhline(thresh, color='red', linestyle='--', label="Threshold")
plt.scatter(hours[alerts], vibration[alerts], color='red', label="Alerts")
plt.title("Predictive Maintenance Threshold")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```
