### Loading CSV Data

There are a number of considerations when loading your machine learning data from CSV files.
For reference, you can learn a lot about the expectations for CSV files by reviewing the CSV
request for comment titled [Common Format and MIME Type for Comma-Separated Values
(CSV) Files](https://tools.ietf.org/html/rfc4180).

#### File Header
Does your data have a file header? If so this can help in automatically assigning names to each
column of data. If not, you may need to name your attributes manually. Either way, you should
explicitly specify whether or not your CSV file had a file header when loading your data.

#### Comments
Does your data have comments? Comments in a CSV file are indicated by a hash (#) at the
start of a line. If you have comments in your file, depending on the method used to load your
data, you may need to indicate whether or not to expect comments and the character to expect
to signify a comment line.

#### Delimiter
The standard delimiter that separates values in fields is the comma (,) character. Your file could
use a different delimiter like tab or white space in which case you must specify it explicitly.

#### Quotes
Sometimes field values can have spaces. In these CSV files the values are often quoted. The
default quote character is the double quotation marks character. Other characters can be used,
and you must specify the quote character used in your file.

### Hull Corrosion Detection Dataset

#### CSV Structure: Hull Corrosion Detection

Hull Corrosion Detection has 768 rows and 9 columns (8 characteristics + 1 target column). It is a CSV file with data for **`corrosion_detection_ships.csv`** dataset. Here’s the description of each column:  

- **`Years_since_last_inspection`**: Number of years since the ship's last hull inspection.  
- **`Water_salinity_PSU`**: Salinity of the water in Practical Salinity Units (PSU), a measure of salt concentration.  
- **`Water_pressure_kPa`**: Water pressure measured in kilopascals (kPa).  
- **`Coating_thickness_mm`**: Thickness of the protective coating on the hull, measured in millimeters (mm).  
- **`Anti_corrosion_concentration_mg_L`**: Concentration of anti-corrosion chemicals in the surrounding water, measured in milligrams per liter (mg/L).  
- **`Hull_material_density_g_cm3`**: Density of the hull's material, measured in grams per cubic centimeter (g/cm³).  
- **`Inspection_history`**: A metric indicating the frequency and quality of previous inspections.  
- **`Ship_age`**: The age of the ship in years.  
- **`Hull_condition`**: Binary indicator representing the hull's condition (0 = good condition, 1 = corroded or damaged).  



### Load CSV Files with the CSV Python Standard Library

The Python API provides the module CSV and the function reader() that can be used to load
CSV files. Once loaded, you can convert the CSV data to a NumPy array and use it for machine
learning. For example, you can download3 the Pima Indians dataset into your local directory
with the filename `Corrosion_Detection_Ships.csv`. All fields in this dataset are numeric
and there is no header line.

The example loads an object that can iterate over each row of the data and can easily be
converted into a NumPy array. Running the example prints the shape of the array.

For more information on the csv.reader() function, see [CSV File Reading and Writing](https://docs.python.org/2/library/csv.html) in
the Python API documentation .

```python
# Load CSV Using CSV Python Standard Library
import csv
import numpy
filename = '/content/sample_data/corrosion_detection_ships.csv'
# Open the file in text mode ('r') instead of binary mode ('rb')
raw_data = open(filename, 'r')
reader = csv.reader(raw_data, delimiter=',', quoting=csv.QUOTE_NONE)
x = list(reader)
data = numpy.array(x).astype('float')
print(data.shape)
```

Correlations Between Attributes

---

Correlation refers to the relationship between two variables and how they may or may not
change together. The most common method for calculating correlation is _**Pearson’s Correlation
Coefficient**_, that assumes a normal distribution of the attributes involved. A correlation of -1
or 1 shows a full negative or positive correlation respectively. Whereas a value of 0 shows no
correlation at all.

Some machine learning algorithms like _**linear and logistic regression**_ can suffer
poor performance if there are highly correlated attributes in your dataset. As such, it is a good
idea to review all of the pairwise correlations of the attributes in your dataset. You can use the
`corr()` function on the Pandas DataFrame to calculate a correlation matrix.

```python
# Pairwise Pearson correlations
from pandas import read_csv
from pandas import set_option
filename = '/content/sample_data/corrosion_detection_ships.csv'
names = ['Plate_Thickness',
'Salinity',
'Temperature',
'PH_Water',
'Exposure_Hours',
'Internal_Pressure',
'Speed',
'Vibrations',
'Critical_Corrosion']
data = read_csv(filename, names=names)
set_option('display.width', 100)
# Set display precision for floats to 3
set_option('display.precision', 3) # Changed 'precision' to 'display.precision'
correlations = data.corr(method='pearson')
print(correlations)
```

The matrix lists all attributes across the top and down the side, to give correlation between
all pairs of attributes (twice, because the matrix is symmetrical). You can see the diagonal
line through the matrix from the top left to bottom right corners of the matrix shows perfect
correlation of each attribute with itself.

Skew of Univariate Distributions

---

Skew refers to a distribution that is assumed Gaussian (normal or bell curve) that is shifted or
squashed in one direction or another. _**`Many machine learning algorithms assume a Gaussian
distribution`**_. Knowing that an attribute has a skew may allow you to perform data preparation
to correct the skew and later improve the accuracy of your models. You can calculate the skew
of each attribute using the skew() function on the Pandas DataFrame.

```python
# Skew for each attribute
from pandas import read_csv
filename = '/content/sample_data/corrosion_detection_ships.csv'
names = ['Plate_Thickness',
'Salinity',
'Temperature',
'PH_Water',
'Exposure_Hours',
'Internal_Pressure',
'Speed',
'Vibrations',
'Critical_Corrosion']
data = read_csv(filename, names=names)
skew = data.skew()
print(skew)
```

The skew result show a positive (right) or negative (left) skew. Values closer to zero show
less skew.

Histograms

---

```python
# Univariate Histograms
from matplotlib import pyplot
from pandas import read_csv
filename = '/content/sample_data/corrosion_detection_ships.csv'
names = ['Plate_Thickness',
'Salinity',
'Temperature',
'PH_Water',
'Exposure_Hours',
'Internal_Pressure',
'Speed',
'Vibrations',
'Critical_Corrosion']
data = read_csv(filename, names=names)
data.hist()
pyplot.show()
```

From the graph, a few things can be observed about the distributions of the data:

#### **Analysis of Distributions:**
1. **Plate_Thickness, Salinity, and Temperature:**  
   - They have fairly uniform distributions, without a clear bell-shaped (Gaussian) or exponential decay form.  
   - This suggests a **uniform distribution** or a combination of several distributions.  

2. **Water_PH and Internal_Pressure:**  
   - They show some bell shape but seem to have skewness.  
   - They could approximate **skewed Gaussian distributions**.  

3. **Exposure_Hours:**  
   - It presents irregular peaks, which could indicate categorical or discrete data rather than a clear continuous distribution.  

4. **Speed and Vibrations:**  
   - They appear to have almost uniform distributions but with fluctuations.  
   - They do not seem to follow a Gaussian or exponential distribution.  

5. **Critical_Corrosion:**  
   - It is clearly **binary** (0 or 1), as expected for the target column.  

#### **How to Handle These Distributions in Machine Learning?**
- **Normalization or Standardization:**  
   Since the distributions are not Gaussian, techniques such as **Min-Max Scaling** or **Standardization (z-score)** can help improve model performance.  

- **Models Robust to Distributions:**  
   - **Decision Trees** and **Random Forest** perform well with non-Gaussian distributions.  
   - **Neural Networks** can adapt but will require normalization.  

- **Variable Transformations:**  
   - For skewed variables, transformations like **logarithmic** or **square root** can be applied.  
   - Uniform variables may benefit from techniques such as **discretization** for certain models.

Correlation Matrix Plot

---

Correlation gives an indication of _**`how related the changes are between two variables`**_. If two
variables change in the same direction they are positively correlated. If they change in opposite
directions together (one goes up, one goes down), then they are negatively correlated.

You can
calculate the correlation between each pair of attributes. This is called a correlation matrix. You
can then plot the _**`correlation matrix`**_ and get an idea of which variables have a high correlation with each other.

This is useful to know, because some machine learning algorithms like _**`linear
and logistic regression`**_ can have _**`poor performance`**_ if there are _**`highly correlated input variables`**_
in your data.

```python
# Correction Matrix Plot
from matplotlib import pyplot
from pandas import read_csv
import numpy
filename = '/content/sample_data/corrosion_detection_ships.csv'
names = ['Plate_Thickness',
'Salinity',
'Temperature',
'PH_Water',
'Exposure_Hours',
'Internal_Pressure',
'Speed',
'Vibrations',
'Critical_Corrosion']
data = read_csv(filename, names=names)
correlations = data.corr()
# plot correlation matrix
fig = pyplot.figure()
ax = fig.add_subplot(111)
cax = ax.matshow(correlations, vmin=-1, vmax=1)
fig.colorbar(cax)
ticks = numpy.arange(0,9,1)
ax.set_xticks(ticks)
ax.set_yticks(ticks)
ax.set_xticklabels(names)
ax.set_yticklabels(names)
pyplot.show()
```

We can see that the matrix is symmetrical, i.e. the bottom left of the matrix is the same as
the top right. This is useful as we can see two different views on the same data in one plot. We
can also see that each variable is perfectly positively correlated with each other (as you would
have expected) in the diagonal line from top left to bottom right.


The example is not generic in that it specifies the names for the attributes along the axes as
well as the number of ticks. This recipe cam be made more generic by removing these aspects
as follows:

---

A scatter plot shows the relationship between two variables as dots in two dimensions, one
axis for each attribute.

You can create a scatter plot for each pair of attributes in your data.
Drawing all these scatter plots together is called a scatter plot matrix. Scatter plots are useful
for spotting structured relationships between variables, like whether you could summarize the
relationship between two variables with a line.

Attributes with structured relationships may
also be correlated and good candidates for removal from your dataset.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler

# Generar dataset original simulando las estadísticas del Pima Indians
def generate_pima_like_data(size=768):
    np.random.seed(42)
    data = {
        'Pregnancies': np.random.poisson(3, size),
        'Glucose': np.random.normal(121, 30, size),
        'BloodPressure': np.random.normal(69, 20, size),
        'SkinThickness': np.random.normal(29, 10, size),
        'Insulin': np.random.normal(80, 115, size).clip(14, 846),
        'BMI': np.random.normal(32, 7, size),
        'DiabetesPedigreeFunction': np.random.normal(0.47, 0.33, size),
        'Age': np.random.normal(33, 11, size).clip(21, 81),
        'Outcome': np.random.choice([0, 1], size=size)
    }
    return pd.DataFrame(data)

# Generar dataset con estadísticas similares
pima_data = generate_pima_like_data()

# Transformaciones para adaptar a Hull Corrosion Detection
df_hull = pd.DataFrame({
    'Years_since_last_inspection': pima_data['Pregnancies'],
    'Water_salinity_PSU': MinMaxScaler((0, 40)).fit_transform(pima_data[['Glucose']])[:, 0], # Extract the first (and only) column
    'Water_pressure_kPa': pima_data['BloodPressure'] * 8,
    'Coating_thickness_mm': pima_data['SkinThickness'] / 50,
    'Anti_corrosion_concentration_mg_L': np.log(pima_data['Insulin']) * 40,
    'Hull_material_density_g_cm3': pima_data['BMI'] * 0.12,
    'Inspection_history': pima_data['DiabetesPedigreeFunction'] * 4,
    'Ship_age': pima_data['Age'],
    'Hull_condition': pima_data['Outcome']
})

# Dibujar histogramas comparativos
def plot_histograms(original_data, transformed_data):
    fig, axes = plt.subplots(2, len(original_data.columns), figsize=(20, 10))

    for i, column in enumerate(original_data.columns):
        axes[0, i].hist(original_data[column], bins=20, alpha=0.7, color='blue')
        axes[0, i].set_title(f'Pima - {column}')

        axes[1, i].hist(transformed_data.iloc[:, i], bins=20, alpha=0.7, color='green')
        axes[1, i].set_title(f'Hull - {transformed_data.columns[i]}')

    plt.tight_layout()
    plt.show()

plot_histograms(pima_data, df_hull)

# Mostrar estadísticas básicas para verificar la coherencia
df_hull.describe().round(2)
```
