## Homework 5 — Techniques from `NB07`-`NB17` most likely relevant to my project

- **Feature engineering fundamentals (`NB03`/`NB04`)**: `WDIR` is genuinely circular (values near 0 and 360 are the same real direction), so a `sin`/`cos` encoding — not a raw degree value — is the correct representation, the same lesson as `NB03`'s dtype/array-mechanics care about not trusting a numeric encoding blindly.
- **Regression fundamentals and metrics (`NB07`)**: `WVHT` is a continuous target, so this is a regression problem scored by RMSE, exactly `NB07`'s Part 3 metric — and `NB07`'s data-leakage lesson applies directly here too (the two wave-derived columns excluded from this dataset are exactly the kind of feature that would leak the answer).
- **Ensembles (`NB08`/`NB10`)**: Random Forest and Gradient Boosting are natural first candidates for a small-to-medium tabular regression problem like this one, the same reasoning `NB08` and `NB10` used for sonar classification and yacht-resistance regression respectively.
- **`NB17`'s decision framework**: labels are known (supervised), the data is tabular (not image/sequence-shaped in the way `NB13`/`NB14` needed), and the dataset size (tens of thousands of rows) sits in the range where `NB17` found classical ML competitive with or better than a neural network — so I would reach for classical ML first, the same way `NB18` and `NB20` did for their own Block 4 case studies, and only try an `NB11`-style MLP as a genuine, open comparison rather than assuming it would win.
- **`NB14`'s time-series framing**: even though I ultimately used point-in-time features rather than sequences, `NB14`'s core lesson — that a naive baseline (predicting the recent mean) is a real, meaningful comparison, not just a formality — is exactly what Checkpoint 2's baseline RMSE is for here.
- **`NB18`'s temporal-holdout discipline**: the project's real evaluation (train on 2019-2023, predict 2024) is the same genuine future-year holdout `NB18` used for its own wind-to-wave-height case study, and `NB18`'s "hold out a year, not a season, when the target is seasonal" principle is exactly why a random or single-season split would have misled me here.
