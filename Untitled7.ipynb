```python
import numpy as np
import matplotlib.pyplot as plt

class SailingSimulator:
    def __init__(self, start=None, target=None):
        self.boat_angle = 0  # Dirección del barco
        self.sail_angle = 0  # Ángulo de la vela respecto al barco
        self.rudder_angle = 0  # Ángulo del timón
        self.position = np.array(start if start is not None else np.random.uniform(-10, 10, size=2))
        self.target = np.array(target if target is not None else np.random.uniform(10, 20, size=2))
        self.velocity = np.array([0.0, 0.0])  # Velocidad del barco
        self.time_elapsed = 0  # Tiempo transcurrido en la simulación
        self.tack_direction = 1  # Dirección del zigzag

    def get_wind_conditions(self, position):
        """Genera un viento variable según la posición del barco."""
        x, y = position
        wind_speed = 8 + 4 * np.sin(0.1 * x) * np.cos(0.1 * y)  # Intensidad variable entre 4 y 12
        wind_angle = np.radians(30 + 20 * np.sin(0.05 * y))  # Dirección variable entre 10° y 50°
        return wind_speed, wind_angle

    def update(self, dt=1.0):
        """Actualiza el estado del barco considerando viento variable."""
        wind_speed, wind_angle = self.get_wind_conditions(self.position)
        wind_force = wind_speed * np.array([np.cos(wind_angle), np.sin(wind_angle)])
        sail_force = wind_force * np.cos(np.radians(self.sail_angle) - wind_angle)
        rudder_effect = np.radians(self.rudder_angle) * 0.1
        water_resistance = np.linalg.norm(self.velocity) * 0.05
        self.boat_angle += rudder_effect - water_resistance
        direction = np.array([np.cos(self.boat_angle), np.sin(self.boat_angle)])
        self.velocity = sail_force * direction * 0.05
        self.position += self.velocity * dt
        self.time_elapsed += dt

    def control_tacking(self):
        """Implementa la estrategia de ceñida para navegar contra el viento."""
        _, wind_angle = self.get_wind_conditions(self.position)
        wind_angle_deg = np.degrees(wind_angle) % 360
        direction_to_target = np.degrees(np.arctan2(self.target[1] - self.position[1],
                                                    self.target[0] - self.position[0])) % 360
        angle_diff = (direction_to_target - wind_angle_deg + 360) % 360

        if angle_diff < 90 or angle_diff > 270:
            self.sail_angle = np.clip(wind_angle_deg / 2, -45, 45)
            self.rudder_angle = 0
        else:
            tack_angle = 45 * self.tack_direction
            self.sail_angle = np.clip(tack_angle, -45, 45)
            self.rudder_angle = np.clip((direction_to_target - (wind_angle_deg + tack_angle)) * 0.3, -25, 25)
            if np.random.rand() < 0.01:
                self.tack_direction *= -1

    def simulate(self, report_interval=60):
        """Ejecuta la simulación con la estrategia de ceñida hasta que el barco llegue al destino."""
        positions = [self.position.copy()]
        next_report_time = report_interval
        while np.linalg.norm(self.position - self.target) > 0.5:
            self.control_tacking()
            self.update()
            positions.append(self.position.copy())
            if self.time_elapsed >= next_report_time:
                distance_to_target = np.linalg.norm(self.position - self.target)
                print(f"Tiempo: {self.time_elapsed:.0f}s - Distancia al objetivo: {distance_to_target:.2f} unidades")
                next_report_time += report_interval
        print(f"Llegó al destino en {self.time_elapsed:.2f} segundos.")
        return np.array(positions), self.time_elapsed

# Configuración de puntos
start_point = np.random.uniform(-10, 10, size=2)
target_point = np.random.uniform(10, 20, size=2)

# Simulación
sim = SailingSimulator(start=start_point, target=target_point)
sim_positions, time_taken = sim.simulate(report_interval=60)

# Graficar
plt.figure(figsize=(8, 6))
plt.plot(sim_positions[:,0], sim_positions[:,1], label='Con Ceñida')
plt.scatter([start_point[0]], [start_point[1]], color='red', label='Inicio (A)')
plt.scatter([target_point[0]], [target_point[1]], color='green', label='Destino (B)')

# Mapa de viento
x_range = np.linspace(min(sim_positions[:, 0]) - 5, max(sim_positions[:, 0]) + 5, 15)
y_range = np.linspace(min(sim_positions[:, 1]) - 5, max(sim_positions[:, 1]) + 5, 15)
for x in x_range:
    for y in y_range:
        wind_speed, wind_angle = sim.get_wind_conditions([x, y])
        wind_vector = np.array([np.cos(wind_angle), np.sin(wind_angle)]) * (wind_speed / 12)
        plt.arrow(x, y, wind_vector[0], wind_vector[1], head_width=0.2, head_length=0.2, fc='blue', ec='blue', alpha=0.6)

plt.xlabel("Posición X")
plt.ylabel("Posición Y")
plt.legend()
plt.title(f"Trayectoria con ceñida (Tiempo a B: {time_taken:.2f}s)")
plt.show()
```
