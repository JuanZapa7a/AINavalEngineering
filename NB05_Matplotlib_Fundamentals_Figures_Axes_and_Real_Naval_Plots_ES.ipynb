## 2. El modelo de objetos Figure/Axes

Una **`Figure`** de Matplotlib es toda la ventana o imagen guardada; un **`Axes`** es un gráfico concreto que vive dentro de ella (a pesar del nombre confusamente parecido, un `Axes` no es lo mismo que las líneas del eje x/y). Una figura puede contener varios axes (Sección 5). El patrón recomendado — usado en todo este curso a partir de aquí — es crear ambos explícitamente con `plt.subplots()` y llamar a los métodos directamente sobre el objeto `ax`, en vez del estilo más antiguo de estado global `plt.plot()`/`plt.xlabel()`, `que se vuelve ambiguo en el momento en que una figura tiene más de una subgráfica`.

```python
import matplotlib.pyplot as plt
import urllib.request
import csv

fuel_url = "https://raw.githubusercontent.com/JuanZapa7a/AINavalEngineering/main/Datasets/ship_fuel_efficiency.csv"
rows = list(csv.DictReader(urllib.request.urlopen(fuel_url).read().decode("utf-8").splitlines()))

distance = [float(r["distance"]) for r in rows]
fuel = [float(r["fuel_consumption"]) for r in rows]

fig, ax = plt.subplots(figsize=(7, 5))   # fig = the whole figure, ax = the one plot inside it
ax.scatter(distance, fuel, s=8, alpha=0.5)
ax.set_xlabel("Distance (nautical miles)")
ax.set_ylabel("Fuel consumption")
ax.set_title(f"Real data: {len(rows)} voyages from ship_fuel_efficiency.csv")
plt.show()
```

> **Para saber más**: [guía rápida de Figure/Axes de Matplotlib](https://matplotlib.org/stable/users/explain/quick_start.html) | [documentación de `matplotlib.pyplot.subplots`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.subplots.html)

## 3. Estilo: color, tipo de línea, marcador, etiquetas

Las 12 lecturas mensuales reales de `fuel_consumption` de un buque (ya usadas así en la predicción con LSTM de `NB14`) permiten construir un gráfico de líneas real limpio y estilizarlo deliberadamente: un color concreto, una línea discontinua, marcadores circulares en cada punto de dato real, y etiquetas de ejes legibles.

```python
ship_id = "NG001"
ship_rows = [r for r in rows if r["ship_id"] == ship_id]   # already in Jan-Dec order in the real file
months = [r["month"][:3] for r in ship_rows]
ship_fuel = [float(r["fuel_consumption"]) for r in ship_rows]

fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(months, ship_fuel, color="darkorange", linestyle="--", marker="o", markersize=6, linewidth=1.5)
ax.set_xlabel("Month")
ax.set_ylabel("Fuel consumption")
ax.set_title(f"Real monthly fuel consumption -- ship {ship_id}")
ax.grid(alpha=0.3)
plt.show()
```

**Pruébalo tú mismo**: reestiliza el mismo gráfico con un color distinto, una línea de puntos, marcadores triangulares (`marker="^"`), y una rejilla suave (`ax.grid(alpha=0.3)`) — cambios pequeños, pero fíjate en cuánto "se lee" de forma distinta cada versión aunque sean los mismos datos reales.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(months, ship_fuel, color="seagreen", linestyle=":", marker="^", markersize=7, linewidth=1.5)
ax.set_xlabel("Month")
ax.set_ylabel("Fuel consumption")
ax.set_title(f"Same real data, restyled -- ship {ship_id}")
ax.grid(alpha=0.3)
plt.show()
```

> **Para saber más**: [colores con nombre de Matplotlib](https://matplotlib.org/stable/gallery/color/named_colors.html) | [estilos de línea/marcador de Matplotlib](https://matplotlib.org/stable/api/markers_api.html)

## 4. Leyendas y anotaciones

Un gráfico con más de una serie necesita una **leyenda** para seguir siendo legible, y una **anotación** puede señalar directamente un valor real concreto que merezca destacarse — aquí, `comparando el consumo mensual de combustible real de tres buques y marcando la única lectura real más alta entre ellos`.

```python
fig, ax = plt.subplots(figsize=(9, 5))

max_point = (None, None, None, -1)  # ship_id, month_label, value, index -- track the real overall max
for sid in ["NG001", "NG002", "NG003"]:
    s_rows = [r for r in rows if r["ship_id"] == sid]
    s_months = [r["month"][:3] for r in s_rows]
    s_fuel = [float(r["fuel_consumption"]) for r in s_rows]
    ax.plot(s_months, s_fuel, marker="o", markersize=4, label=sid)
    local_max_idx = s_fuel.index(max(s_fuel))
    if max_point[2] is None or s_fuel[local_max_idx] > max_point[2]:
        max_point = (sid, s_months[local_max_idx], s_fuel[local_max_idx], local_max_idx)

ax.annotate(
    f"Highest reading: {max_point[0]}, {max_point[1]} ({max_point[2]:.0f})",
    xy=(max_point[3], max_point[2]),
    xytext=(max_point[3] + 1, max_point[2] + 300),
    arrowprops=dict(arrowstyle="->", color="black"),
)
ax.set_xlabel("Month")
ax.set_ylabel("Fuel consumption")
ax.set_title("Real monthly fuel consumption -- 3 ships")
ax.legend(title="Ship ID")
plt.show()
```

> **Para saber más**: [guía de leyendas de Matplotlib](https://matplotlib.org/stable/users/explain/axes/legend_guide.html) | [guía de anotaciones de Matplotlib](https://matplotlib.org/stable/users/explain/text/annotations.html)

## 5. Disposiciones de subgráficas

`plt.subplots(rows, cols)` devuelve toda una rejilla de `Axes` de una vez. La siguiente celda construye un panel real 2x2 a partir del dataset de combustible: un histograma, un diagrama de caja por tipo de buque, un diagrama de dispersión, y un diagrama de barras con las medias por tipo — `cuatro vistas reales distintas del mismo dataset en una sola figura`.

```python
import numpy as np

ship_types = sorted(set(r["ship_type"] for r in rows))
fuel_by_type = {t: [float(r["fuel_consumption"]) for r in rows if r["ship_type"] == t] for t in ship_types}

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

axes[0, 0].hist(fuel, bins=25, color="steelblue", edgecolor="white")
axes[0, 0].set_title("Distribution of fuel consumption")

axes[0, 1].boxplot([fuel_by_type[t] for t in ship_types], tick_labels=ship_types)
axes[0, 1].set_title("Fuel consumption by ship type")
axes[0, 1].tick_params(axis="x", rotation=30)

axes[1, 0].scatter(distance, fuel, s=6, alpha=0.4, color="seagreen")
axes[1, 0].set_title("Distance vs. fuel consumption")
axes[1, 0].set_xlabel("Distance")

axes[1, 1].bar(ship_types, [np.mean(fuel_by_type[t]) for t in ship_types], color="coral")
axes[1, 1].set_title("Average fuel consumption per type")
axes[1, 1].tick_params(axis="x", rotation=30)

fig.suptitle("Four real views of the same dataset, one figure", fontsize=13)
plt.tight_layout()
plt.show()
```

Para disposiciones en las que los paneles no forman una rejilla uniforme (por ejemplo, un gráfico ancho arriba y dos más estrechos debajo), `plt.subplots` por sí solo no puede expresarlo — `matplotlib.gridspec.GridSpec` sí puede.

```python
import matplotlib.gridspec as gridspec

fig = plt.figure(figsize=(9, 6))
gs = gridspec.GridSpec(2, 2, figure=fig)

ax_top = fig.add_subplot(gs[0, :])         # spans both columns, top row
ax_bottom_left = fig.add_subplot(gs[1, 0])
ax_bottom_right = fig.add_subplot(gs[1, 1])

ax_top.plot(months, ship_fuel, marker="o")
ax_top.set_title(f"Wide top panel: ship {ship_id} monthly fuel")

ax_bottom_left.hist(fuel_by_type["Tanker Ship"], bins=15, color="slateblue")
ax_bottom_left.set_title("Tanker Ship distribution")

ax_bottom_right.hist(fuel_by_type["Fishing Trawler"], bins=15, color="darkgoldenrod")
ax_bottom_right.set_title("Fishing Trawler distribution")

plt.tight_layout()
plt.show()
```

> **Para saber más**: [guía de subgráficas de Matplotlib](https://matplotlib.org/stable/users/explain/axes/arranging_axes.html) | [documentación de `matplotlib.gridspec`](https://matplotlib.org/stable/api/_as_gen/matplotlib.gridspec.GridSpec.html)

## 6. Gráficos de mapa de color sobre datos reales

`imshow` representa un array 2-D directamente como una rejilla de color — exactamente la misma llamada de representación gráfica sobre la que se construyen tanto las rejillas ERA5 de `NB18` como la rejilla de batimetría de `NB20`, antes de añadirles líneas de costa geográficas reales con `cartopy`. Aquí visualiza algo ya real y ya calculado en este curso: la matriz de correlación completa de 60x60 entre cada par de bandas de frecuencia del Sonar (`NB08`).

```python
sonar_url = "https://raw.githubusercontent.com/JuanZapa7a/AINavalEngineering/main/Datasets/sonar.all-data"
sonar_lines = urllib.request.urlopen(sonar_url).read().decode("utf-8").strip().split("\n")
sonar_features = np.array([[float(x) for x in line.split(",")[:-1]] for line in sonar_lines])

corr_matrix = np.corrcoef(sonar_features.T)   # real 60x60 correlation matrix

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(corr_matrix, cmap="coolwarm", vmin=-1, vmax=1)
ax.set_xlabel("Frequency band index")
ax.set_ylabel("Frequency band index")
ax.set_title("Real correlation matrix -- 60 Sonar frequency bands")
plt.colorbar(im, ax=ax, label="Correlation")
plt.show()
```

**Pruébalo tú mismo**: la diagonal (`siempre exactamente 1.0`) domina la escala de color aunque no sea la parte interesante. Vuelve a representar la misma matriz de correlación con `cmap="viridis"` y la diagonal enmascarada (ponla a `np.nan` sobre una copia de la matriz primero) — ¿se ve más fácilmente una estructura real distinta?

```python
masked_corr = corr_matrix.copy()
np.fill_diagonal(masked_corr, np.nan)

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(masked_corr, cmap="viridis", vmin=-1, vmax=1)
ax.set_xlabel("Frequency band index")
ax.set_ylabel("Frequency band index")
ax.set_title("Same real correlations, diagonal masked, different colormap")
plt.colorbar(im, ax=ax, label="Correlation")
plt.show()
```

La diagonal brillante (siempre exactamente 1.0 -- cada banda se correlaciona perfectamente consigo misma) y la banda roja amplia cercana a ella (las bandas de frecuencia vecinas se correlacionan fuertemente entre sí) son ambos patrones reales y físicamente esperables: `las frecuencias de sonar adyacentes responden al mismo eco subyacente, así que suben y bajan juntas`.

> **Para saber más**: [documentación de `matplotlib.pyplot.imshow`](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.imshow.html) | [referencia de mapas de color de Matplotlib](https://matplotlib.org/stable/users/explain/colors/colormaps.html)

## 7. Ejes gemelos

Dos magnitudes reales con escalas y unidades genuinamente distintas — `distance` (millas náuticas) y `fuel_consumption` (sus propias unidades) para los 12 meses de un buque real — son difíciles de comparar en un único eje y compartido. `ax.twinx()` crea un segundo eje y que comparte el mismo eje x, de modo que ambas series pueden leerse en su propia escala natural.

```python
ship_distance = [float(r["distance"]) for r in ship_rows]

fig, ax1 = plt.subplots(figsize=(8, 4.5))
ax2 = ax1.twinx()

line1, = ax1.plot(months, ship_distance, color="steelblue", marker="o", label="Distance")
line2, = ax2.plot(months, ship_fuel, color="darkorange", marker="s", label="Fuel consumption")

ax1.set_xlabel("Month")
ax1.set_ylabel("Distance (nm)", color="steelblue")
ax2.set_ylabel("Fuel consumption", color="darkorange")
ax1.tick_params(axis="y", labelcolor="steelblue")
ax2.tick_params(axis="y", labelcolor="darkorange")
ax1.set_title(f"Real distance and fuel consumption together -- ship {ship_id}")
ax1.legend(handles=[line1, line2], loc="upper left")
plt.show()
```

**Pruébalo tú mismo**: reconstruye este mismo gráfico de ejes gemelos para un buque real distinto (`ship_id = "NG002"`) — misma estructura, datos reales nuevos. ¿Se mueven la distancia y el consumo de combustible de forma igual de clara en conjunto?

```python
ship_id_2 = "NG002"
ship_rows_2 = [r for r in rows if r["ship_id"] == ship_id_2]
months_2 = [r["month"][:3] for r in ship_rows_2]
fuel_2 = [float(r["fuel_consumption"]) for r in ship_rows_2]
distance_2 = [float(r["distance"]) for r in ship_rows_2]

fig, ax1 = plt.subplots(figsize=(8, 4.5))
ax2 = ax1.twinx()

line1, = ax1.plot(months_2, distance_2, color="steelblue", marker="o", label="Distance")
line2, = ax2.plot(months_2, fuel_2, color="darkorange", marker="s", label="Fuel consumption")

ax1.set_xlabel("Month")
ax1.set_ylabel("Distance (nm)", color="steelblue")
ax2.set_ylabel("Fuel consumption", color="darkorange")
ax1.tick_params(axis="y", labelcolor="steelblue")
ax2.tick_params(axis="y", labelcolor="darkorange")
ax1.set_title(f"Real distance and fuel consumption together -- ship {ship_id_2}")
ax1.legend(handles=[line1, line2], loc="upper left")
plt.show()
```

## 8. Resaltar un rango con `fill_between`

A veces lo que importa no es una única línea sino un *rango* — la dispersión entre todos los buques en un mes dado, no solo el valor de un buque. `ax.fill_between()` sombrea el área entre dos límites (aquí, el mínimo y el máximo reales de toda la flota) justo debajo de una línea, haciendo ese rango visible de inmediato en vez de obligar al lector a comparar gráficos separados.

```python
month_labels = [r["month"][:3] for r in ship_rows]   # Jan..Dec, already in order
month_full = [r["month"] for r in ship_rows]

monthly_mean, monthly_min, monthly_max = [], [], []
for m in month_full:
    vals = [float(r["fuel_consumption"]) for r in rows if r["month"] == m]
    monthly_mean.append(np.mean(vals))
    monthly_min.append(np.min(vals))
    monthly_max.append(np.max(vals))

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(month_labels, monthly_mean, color="steelblue", marker="o", label="Fleet mean")
ax.fill_between(month_labels, monthly_min, monthly_max, alpha=0.2, color="steelblue", label="Fleet min-max range")
ax.set_xlabel("Month")
ax.set_ylabel("Fuel consumption")
ax.set_title("Real fleet-wide fuel consumption range, by month")
ax.legend()
plt.show()
```

> **Para saber más**: [documentación de `matplotlib.axes.Axes.fill_between`](https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.fill_between.html)

## 9. Guardar figuras

`fig.savefig(...)` escribe la figura real en un archivo con una resolución (`dpi`) y un formato elegidos (PNG, PDF, SVG...). Esto importa en el momento en que un gráfico tiene que ir a un informe y no solo a una celda del notebook — `la resolución en pantalla por defecto suele ser demasiado baja para imprimir`.

```python
fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(months, ship_fuel, marker="o", color="darkorange")
ax.set_title(f"Ship {ship_id} -- saved at publication quality")
fig.savefig("ship_fuel_plot.png", dpi=200, bbox_inches="tight")
print("Saved ship_fuel_plot.png")
```

> **Para saber más**: [documentación de `Figure.savefig`](https://matplotlib.org/stable/api/_as_gen/matplotlib.figure.Figure.savefig.html)

## Resumen de la clase

- Construye todo gráfico real mediante `fig, ax = plt.subplots()`, no con el estilo más antiguo de estado global `plt.plot()` — se mantiene sin ambigüedad en el momento en que una figura tiene más de una subgráfica.
- Un estilo deliberado (color, tipo de línea, marcador) y las etiquetas hacen que un gráfico se explique por sí solo, en vez de depender de un pie de figura que lo explique.
- Las leyendas y las anotaciones son lo que mantiene legible un gráfico con varias series o con un único valor destacado.
- `plt.subplots(rows, cols)` cubre rejillas uniformes; `GridSpec` cubre disposiciones que no lo son.
- `imshow` convierte cualquier array 2-D real en un gráfico de mapa de color — la base directa de los mapas geográficos de `NB18` y `NB20`.
- Los ejes gemelos permiten que dos magnitudes reales con unidades distintas compartan honestamente un mismo eje x, cada una en su propia escala.
