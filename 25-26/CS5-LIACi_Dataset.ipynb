# The LIACI Dataset


The **LIACi dataset** (Lifecycle Inspection, Analysis and Condition information), developed by **SINTEF Ocean** in Norway, is a pioneering resource in the field of **computer vision for underwater ship hull inspections**. ([Underwater-ship-hull-Inspection-using-LIACi-dataset/data ... - GitHub](https://github.com/AbubakarAliyuBadawi/Underwater-ship-hull-Inspection-using-LIACi-dataset/blob/master/data_processing.py?utm_source=chatgpt.com) ["GitHub - SINTEF/LIACI_context: Fusion of Multi-Modal Data from ..."](https://github.com/SINTEF/LIACI_context?utm_source=chatgpt.com))


**Main Features of the LIACi Dataset are:**

* **Content**:
  It includes **1,893 real underwater inspection images**, each annotated at the **pixel level** by human experts. (["Semantic Segmentation in Underwater Ship Inspections: Benchmark and ..."](https://www.sintef.no/en/publications/publication/2071613/?utm_source=chatgpt.com))

* **Semantic Classes**:
  The images are labeled with **11 categories**, including (["Semantic Segmentation in Underwater Ship Inspections: Benchmark and ..."](https://www.sintef.no/en/publications/publication/2071613/?utm_source=chatgpt.com)):

  * Ship hull
  * Marine growth
  * Anodes
  * Discharge valves
  * Propellers
  * Paint flaking
  * Bilge keel
  * Defects
  * Corrosion
  * Sea chest gratings
  * Background

* **License and Access**:
  The dataset is available for **non-commercial use** under the **Creative Commons Attribution-NonCommercial-ShareAlike 4.0 International** license.
  It can be accessed at [liaci.sintef.cloud](https://liaci.sintef.cloud), although registration with a Google account is required. (["LIACi - Lifecycle Inspection, Analysis and Condition ... - SINTEF"](https://www.sintef.no/en/projects/2021/liaci/?utm_source=chatgpt.com))


**Some Applications and Associated Tools used with LIACI:**

* **Underwater Semantic Segmentation**:
  LIACi has been used to train semantic segmentation models such as **U-Net with MobileNetV2**, achieving a balance between **accuracy and computational efficiency**. (["Semantic Segmentation in Underwater Ship Inspections: Benchmark and Dataset"](https://ntnuopen.ntnu.no/ntnu-xmlui/handle/11250/3052506?utm_source=chatgpt.com))

* **Open-world Segmentation**:
  Approaches have been developed using LIACi for **open-world semantic segmentation**, capable of identifying both **known classes** and **previously unseen anomalies**.
  One of these projects uses a **dual-decoder architecture with a ResNet34 backbone**, achieving a mean **IoU of 50.9% across 11 classes**. (["GitHub - Ishfaz/Open-world-Semantic-Segmentation-for-Underwater ..."](https://github.com/Ishfaz/Open-world-Semantic-Segmentation-for-Underwater-Shiphull-Inspection-on-LIACi-dataset?utm_source=chatgpt.com))

* **Multimodal Data Contextualization**:
  The LIACi project has also explored the **fusion of multimodal underwater inspection data** using **knowledge graphs**, improving decision support systems. (["GitHub - SINTEF/LIACI_context: Fusion of Multi-Modal Data from ..."](https://github.com/SINTEF/LIACI_context?utm_source=chatgpt.com))


**Some Additional Resources are:**

* **Interactive Demonstrators**:
  SINTEF has developed demonstration applications that allow visualization of **underwater video quality** and **detection of holes in aquaculture nets**, based on LIACi data. (["LIACi - Lifecycle Inspection, Analysis and Condition ... - SINTEF"](https://www.sintef.no/en/projects/2021/liaci/?utm_source=chatgpt.com))

* **Code Repository**:
  The **source code** related to data contextualization and other tools is available on GitHub, facilitating **reproducibility and extension** of the work. (["GitHub - SINTEF/LIACI_context: Fusion of Multi-Modal Data from ..."](https://github.com/SINTEF/LIACI_context?utm_source=chatgpt.com))

```python
import os
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# Choose an image
img_index = 'image_0001.jpg'
image = Image.open(os.path.join(imgs_path, img_index)).convert('RGB')
image_np = np.array(image)

# Get list of classes and assign a unique color to each
mask_classes = sorted(os.listdir(masks_path))
colors = plt.colormaps.get_cmap('tab10').resampled(len(mask_classes)) # 10-class color map


# Create overlay image (copy of original)
overlay = image_np.copy()

# Create a label mask to show legend later
legend_info = []

for i, cls in enumerate(mask_classes):
    mask_file = os.path.join(masks_path, cls, img_index.replace('.jpg', '.bmp'))
    if os.path.exists(mask_file):
        mask = Image.open(mask_file).convert('L')
        mask_np = np.array(mask)

        # Create boolean mask (mask > 0)
        binary_mask = mask_np > 0

        # Apply color with alpha blending
        color = np.array(colors(i)[:3]) * 255  # RGB from colormap
        overlay[binary_mask] = (
            0.5 * overlay[binary_mask] + 0.5 * color
        ).astype(np.uint8)

        legend_info.append((cls, color.astype(int)))

# Plot the overlaid image
plt.figure(figsize=(4*1.7, 3*1.7))
plt.imshow(overlay)
plt.title(f"Overlay of Masks on {img_index}")
plt.axis('off')

# Show legend
handles = [plt.Rectangle((0,0),1,1, color=np.array(c)/255) for _, c in legend_info]
labels = [name for name, _ in legend_info]
plt.legend(handles, labels, bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

masks_combined/ folder contains a single segmentation map per image, where each pixel’s value represents a semantic class index.

```python
import os
import numpy as np
from PIL import Image

# Add path
# base_dir = 'Datasets/LIACi_dataset_pretty'
# images_dir = os.path.join(base_dir, 'images')
# masks_dir = os.path.join(base_dir, 'masks')
masks_c_path = os.path.join(base_dir, 'masks_combined')
os.makedirs(masks_c_path, exist_ok=True)

# List of class folders
mask_classes = sorted(os.listdir(masks_path))
class_to_index = {cls: idx + 1 for idx, cls in enumerate(mask_classes)}  # 0 = background

print(f"Class to index mapping:\n{class_to_index}")

# Process all images
for image_name in sorted(os.listdir(imgs_path)):
    if not image_name.endswith('.jpg'):
        continue

    # Load base image to get size
    image_path = os.path.join(imgs_path, image_name)
    with Image.open(image_path) as img:
        width, height = img.size

    seg_map = np.zeros((height, width), dtype=np.uint8)

    for cls_name in mask_classes:
        cls_index = class_to_index[cls_name]
        mask_file = os.path.join(masks_path, cls_name, image_name.replace('.jpg', '.bmp'))

        if os.path.exists(mask_file):
            mask = Image.open(mask_file).convert('L')
            mask_np = np.array(mask) > 0
            seg_map[mask_np] = cls_index  # overwrite with class index

    # Save segmentation map
    combined_mask = Image.fromarray(seg_map)
    combined_mask.save(os.path.join(masks_c_path, image_name.replace('.jpg', '.png')))

print("All combined masks saved in:", masks_c_path)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from random import sample
import matplotlib.patches as mpatches

# Paths
# image_dir = 'Datasets/LIACi_dataset_pretty/images'
# mask_dir = 'Datasets/LIACi_dataset_pretty/masks_combined'
# masks_root = 'Datasets/LIACi_dataset_pretty/masks'

# Class mappings
mask_classes = sorted(os.listdir(masks_path))
class_to_index = {cls: idx + 1 for idx, cls in enumerate(mask_classes)}  # 0 = background
index_to_class = {0: 'background'}
for cls, idx in class_to_index.items():
    index_to_class[idx] = cls

# Colormap
cmap = plt.get_cmap('tab20')

# Sample files
image_files = sorted(os.listdir(imgs_path))
sample_files = sample(image_files, 3)

for fname in sample_files:
    # Load image and mask
    image = Image.open(os.path.join(imgs_path, fname)).convert('RGB')
    mask = Image.open(os.path.join(masks_c_path, fname.replace('.jpg', '.png')))
    mask_np = np.array(mask)
    image_np = np.array(image)

    # Convert image to RGBA
    base_img_rgba = image.convert("RGBA")

    # Initialize transparent overlay image as PIL.Image in RGBA
    overlay_rgba = Image.new("RGBA", base_img_rgba.size, (0, 0, 0, 0))
    alpha_per_class = 100  # transparency (0–255)

    # Loop through classes and alpha composite each mask
    for idx in np.unique(mask_np):
        if idx == 0:
            continue  # Skip background

        # Create single-channel binary mask for this class
        class_mask = (mask_np == idx).astype(np.uint8)

        # Create color layer for this class
        color_rgb = (np.array(cmap(idx)[:3]) * 255).astype(np.uint8)
        color_layer = np.zeros((mask_np.shape[0], mask_np.shape[1], 4), dtype=np.uint8)
        color_layer[..., :3] = color_rgb
        color_layer[..., 3] = class_mask * alpha_per_class  # apply alpha only where mask is

        class_layer_pil = Image.fromarray(color_layer, mode='RGBA')
        overlay_rgba = Image.alpha_composite(overlay_rgba, class_layer_pil)

    # Final blend: base image + overlay
    result = Image.alpha_composite(base_img_rgba, overlay_rgba)

    # Plot
    fig, axs = plt.subplots(1, 3, figsize=(6.8,2), gridspec_kw={'width_ratios': [1.2, 1.2, 0.8]})

    axs[0].imshow(image)
    axs[0].set_title(f'Original: {fname}', fontsize=10)
    axs[0].axis('off')

    axs[1].imshow(result)
    axs[1].set_title('Transparent Class Overlays', fontsize=10)
    axs[1].axis('off')

    # Legend
    patches = []
    for idx in np.unique(mask_np):
        class_name = index_to_class.get(idx, f'class_{idx}')
        color = cmap(idx)
        patch = mpatches.Patch(color=color, label=f"{idx}: {class_name}")
        patches.append(patch)

    axs[2].legend(handles=patches, loc='center left', frameon=False)
    axs[2].axis('off')
    axs[2].set_title('Legend', fontsize=10)

    plt.tight_layout()
    plt.show()
```

## Dataset

**Define a PyTorch Dataset**

---

```python
import os
from PIL import Image
import torch
from torch.utils.data import Dataset
import torchvision.transforms as T
import numpy as np # Import numpy here

class LIACiDataset(Dataset):
    def __init__(self, dataframe, image_dir, mask_dir, transform=None):
        self.df = dataframe.reset_index(drop=True)
        self.image_dir = imgs_path
        self.mask_dir = masks_c_path
        self.transform = transform

        self.images = sorted(os.listdir(image_dir))

    def __len__(self):
        return len(self.images)

    def __getitem__(self, idx):
        img_name = self.images[idx]
        img_path = os.path.join(self.image_dir, img_name)
        mask_path = os.path.join(self.mask_dir, img_name.replace('.jpg', '.png'))

        image = Image.open(img_path).convert('RGB')
        mask = Image.open(mask_path)  # Should be 2D with class indices

        if self.transform:
            image = self.transform(image)
            for t in self.transform.transforms:
                if isinstance(t, T.Resize):
                    mask = mask.resize(t.size, Image.Resampling.NEAREST)
                    break
            mask = torch.from_numpy(np.array(mask)).long()
        else:
            mask = torch.from_numpy(np.array(mask)).long()

        return image, mask
```

```python
import segmentation_models_pytorch as smp

# Número de clases (incluyendo fondo)
NUM_CLASSES = len(os.listdir('Datasets/LIACi_dataset_pretty/masks'))+1

# Modelo U-Net con encoder resnet18 (puedes usar 'mobilenet_v2' también)
model = smp.Unet(
    encoder_name='resnet18',        # backbone
    encoder_weights='imagenet',     # pretrained weights
    in_channels=3,                  # RGB images
    classes=NUM_CLASSES             # output classes
)
```

## Train the Model

---
**Train the model**

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from tqdm import tqdm
import matplotlib.pyplot as plt
import numpy as np

# Asegúrate de tener el modelo, dataloaders, criterion y optimizer definidos
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model.to(device)

# Métrica IoU
def compute_iou(pred, mask, num_classes):
    ious = []
    pred = pred.view(-1)
    mask = mask.view(-1)
    for cls in range(num_classes):
        pred_inds = pred == cls
        target_inds = mask == cls
        intersection = (pred_inds[target_inds]).sum().item()
        union = pred_inds.sum().item() + target_inds.sum().item() - intersection
        if union == 0:
            ious.append(float('nan'))  # skip this class
        else:
            ious.append(intersection / union)
    return np.nanmean(ious)

# Entrenamiento
def train_model(model, train_loader, val_loader, criterion, optimizer, num_epochs=10, num_classes=12):
    best_val_iou = 0
    train_losses = []
    val_losses = []
    val_ious = []

    for epoch in range(num_epochs):
        model.train()
        running_loss = 0.0
        for images, masks in tqdm(train_loader, desc=f'Epoch {epoch+1}/{num_epochs} - Training'):
            images = images.to(device)
            masks = masks.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, masks)
            loss.backward()
            optimizer.step()
            running_loss += loss.item() * images.size(0)

        train_loss = running_loss / len(train_loader.dataset)
        train_losses.append(train_loss)

        # Validación
        model.eval()
        val_loss = 0.0
        iou_total = 0
        with torch.no_grad():
            for images, masks in tqdm(val_loader, desc=f'Epoch {epoch+1}/{num_epochs} - Validation'):
                images = images.to(device)
                masks = masks.to(device)

                outputs = model(images)
                loss = criterion(outputs, masks)
                val_loss += loss.item() * images.size(0)

                preds = torch.argmax(outputs, dim=1)
                iou_total += compute_iou(preds.cpu(), masks.cpu(), num_classes)

        val_loss /= len(val_loader.dataset)
        mean_iou = iou_total / len(val_loader)
        val_losses.append(val_loss)
        val_ious.append(mean_iou)

        print(f"Epoch {epoch+1}: Train Loss = {train_loss:.4f} | Val Loss = {val_loss:.4f} | Val IoU = {mean_iou:.4f}")

        # Guardar el mejor modelo
        if mean_iou > best_val_iou:
            best_val_iou = mean_iou
            torch.save(model.state_dict(), 'best_model.pth')
            print(" Best model saved!")

    print(f" Training complete. Best Val IoU: {best_val_iou:.4f}")

    return train_losses, val_losses, val_ious
```

```python
def evaluate(model, dataloader, num_classes, device='cuda'):
    model.eval()
    iou_per_class = np.zeros(num_classes)
    total_per_class = np.zeros(num_classes)
    correct_pixels = 0
    total_pixels = 0

    with torch.no_grad():
        for images, masks in dataloader:
            images = images.to(device)
            masks = masks.to(device)

            outputs = model(images)
            preds = torch.argmax(outputs, dim=1)

            for cls in range(num_classes):
                pred_inds = (preds == cls)
                target_inds = (masks == cls)
                intersection = (pred_inds & target_inds).sum().item()
                union = (pred_inds | target_inds).sum().item()
                if union != 0:
                    iou_per_class[cls] += intersection
                    total_per_class[cls] += union

            correct_pixels += (preds == masks).sum().item()
            total_pixels += masks.numel()

    ious = iou_per_class / (total_per_class + 1e-6)
    miou = np.mean(ious)
    acc = correct_pixels / total_pixels

    print("\n=== Evaluation Metrics ===")
    for idx, iou in enumerate(ious):
        print(f"Class {idx}: IoU = {iou:.4f}")
    print(f"Mean IoU: {miou:.4f}")
    print(f"Pixel Accuracy: {acc:.4f}")

    return ious, miou, acc
```

```python
import os
import numpy as np
from PIL import Image

# Clase -> índice
class_map = {
    'anode': 1,
    'bilge_keel': 2,
    'corrosion': 3,
    'defect': 4,
    'marine_growth': 5,
    'over_board_valves': 6,
    'paint_peel': 7,
    'propeller': 8,
    'saliency': 9,
    'sea_chest_grating': 10,
    'ship_hull': 12  # fondo es 0
}

masks_dir = 'Datasets/LIACi_dataset_pretty/masks'
output_dir = 'Datasets/LIACi_dataset_pretty/masks_combined'
os.makedirs(output_dir, exist_ok=True)

# Asumimos que todos los nombres de imagen están sincronizados
image_list = sorted(os.listdir(os.path.join(masks_dir, 'anode')))

for img_name in image_list:
    h, w = np.array(Image.open(os.path.join(masks_dir, 'anode', img_name))).shape
    combined_mask = np.zeros((h, w), dtype=np.uint8)

    for class_name, class_index in class_map.items():
        mask_path = os.path.join(masks_dir, class_name, img_name)
        if os.path.exists(mask_path):
            mask = np.array(Image.open(mask_path))
            # Asegúrate de que solo sobrescriba fondo (0)
            combined_mask[(mask > 0) & (combined_mask == 0)] = class_index

    Image.fromarray(combined_mask).save(os.path.join(output_dir, img_name.replace('.bmp', '.png')))
```
