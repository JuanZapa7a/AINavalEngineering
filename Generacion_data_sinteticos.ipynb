````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import MinMaxScaler

# --------------------------------------
# 1. Cargar los datasets
# --------------------------------------
# Dataset Pima Indians Diabetes
url_pima = 'https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv'
columns_pima = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
                'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
pima_df = pd.read_csv(url_pima, names=columns_pima)

# Dataset Sintético (generado previamente)
np.random.seed(42)
n = 768
sintetico_df = pd.DataFrame({
    'Num_Inspecciones': np.random.normal(3.8, 3.4, n).clip(0).round(),
    'Salinidad': np.random.normal(120.9, 32.0, n).clip(0),
    'Presion_Agua': np.random.normal(69.1, 19.4, n).clip(0),
    'Espesor_Recubrimiento': np.random.normal(20.5, 16.0, n).clip(0),
    'Nivel_Corrosion': np.random.normal(79.8, 115.2, n).clip(0, 100),
    'Indice_Mantenimiento': np.random.normal(0.5, 0.3, n).clip(0, 1),
    'Edad_Barco': np.random.normal(33.2, 11.8, n).clip(0),
    'Funcion_Riesgo': np.random.normal(0.4719, 0.331, n).clip(0),
    'Resultado': (np.random.normal(79.8, 115.2, n).clip(0, 100) > 50).astype(int)
})

# --------------------------------------
# 2. Normalizar los datos (escala 0-1)
# --------------------------------------
# Variables a comparar (excluyendo las target)
pima_columns = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness',
                'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age']
sintetico_columns = ['Num_Inspecciones', 'Salinidad', 'Presion_Agua',
                     'Espesor_Recubrimiento', 'Nivel_Corrosion',
                     'Indice_Mantenimiento', 'Edad_Barco', 'Funcion_Riesgo']

# Normalización Min-Max
scaler = MinMaxScaler()
pima_normalized = pd.DataFrame(scaler.fit_transform(pima_df[pima_columns]), columns=pima_columns)
sintetico_normalized = pd.DataFrame(scaler.fit_transform(sintetico_df[sintetico_columns]), columns=sintetico_columns)

# --------------------------------------
# 3. Graficar histogramas normalizados
# --------------------------------------
plt.figure(figsize=(14, 10))

# Subplot para Pima Indians
plt.subplot(2, 1, 1)
for column in pima_columns:
    sns.histplot(pima_normalized[column], kde=True, alpha=0.5, label=column)
plt.title('Distribución de variables - Pima Indians (Normalizado 0-1)')
plt.xlabel('Valores normalizados')
plt.ylabel('Frecuencia')
plt.legend(loc='upper right', bbox_to_anchor=(1.25, 1))

# Subplot para Dataset Sintético
plt.subplot(2, 1, 2)
for column in sintetico_columns:
    sns.histplot(sintetico_normalized[column], kde=True, alpha=0.5, label=column)
plt.title('Distribución de variables - Dataset Sintético (Normalizado 0-1)')
plt.xlabel('Valores normalizados')
plt.ylabel('Frecuencia')
plt.legend(loc='upper right', bbox_to_anchor=(1.25, 1))

plt.tight_layout()
plt.show()

# --------------------------------------
# 4. Opcional: Boxplots estandarizados (Z-score)
# --------------------------------------
from sklearn.preprocessing import StandardScaler

# Estandarizar
scaler_std = StandardScaler()
pima_std = pd.DataFrame(scaler_std.fit_transform(pima_df[pima_columns]), columns=pima_columns)
sintetico_std = pd.DataFrame(scaler_std.fit_transform(sintetico_df[sintetico_columns]), columns=sintetico_columns)

# Combinar y graficar
combined_std = pd.concat([pima_std, sintetico_std], axis=1)
plt.figure(figsize=(14, 6))
sns.boxplot(data=combined_std, palette=['blue']*8 + ['orange']*8)
plt.xticks(rotation=45)
plt.title('Comparación de variables estandarizadas (Z-score)')
plt.show()
````

### **¿Dónde tiene sentido?**  
1. **Como ejercicio conceptual**:  
   - Útil para enseñar cómo "traducir" variables entre dominios distintos (ej.: en un curso de ciencia de datos).  
   - Ejemplo:  
     - *"La glucosa en sangre es a la diabetes, como la salinidad del agua es a la corrosión"*.  

2. **Para prototipado rápido**:  
   - Si solo necesitas un dataset *similar en estructura* (8 variables numéricas + 1 binaria) para probar un modelo de clasificación, sirve.  

3. **Analogías interesantes**:  
   - **`DiabetesPedigreeFunction`** → Historial de corrosión en la flota: Ambas son "riesgos heredados".  
   - **`BMI`** → Índice de deterioro: Ambos son métricas compuestas que resumen "salud".  

---

### **¿Dónde NO tiene sentido?**  
1. **Unidades y escalas**:  
   - La **presión arterial** (mmHg) no es equivalente a la **presión del agua en el casco** (kPa).  
   - El **número de embarazos** es discreto, mientras que las **inspecciones críticas** podrían tener decimales.  

2. **Mecanismos físicos vs. biológicos**:  
   - La diabetes es un proceso bioquímico; la corrosión es electroquímica/física. No hay correlación causal real.  

3. **Datos reales**:  
   - En ingeniería naval, variables como **tipo de material**, **velocidad del barco**, o **temperatura del agua** son críticas y no están en el dataset Pima.

### **Conclusión**  
- **✅ Tiene sentido** como analogía teórica o para prototipado.**  
- **❌ No es válido** para análisis técnicos reales en corrosión o mantenimiento naval.**
