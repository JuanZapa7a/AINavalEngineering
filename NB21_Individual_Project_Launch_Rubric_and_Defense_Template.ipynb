## 2. Why defending your own work is a real engineering skill


A model that works but that its own author can't explain is not a deliverable an engineering team, a classifying society, or a client will accept. **"Why did you choose this model?" "How do you know it actually works?" "What would break it?"** are the real questions asked of any engineering analysis before it's trusted — not just in this course, but in a real naval engineering career, whenever a colleague, a surveyor, or a regulator needs to trust a result they didn't produce themselves.

This is also, directly, one of this program's own stated competencies (guía docente `CG08`): "capacidad para el análisis e interpretación de mediciones, cálculos, valoraciones... e informes." A defense is that competency, examined directly.

## 3. The evaluation rubric, in depth


The guía docente states the individual project is graded on "resolución individual y exposición" — resolution *and* exposition, both required. In this course's own established practice (every case study since `NB14`), the qualities that actually distinguish a strong result from a weak one are consistent, and rarely just "the final accuracy number":


```python
rubric = {
    "Real understanding of the data": "Can you explain what every column means, and why you chose your features -- not just that you loaded a CSV?",
    "A genuine, appropriate evaluation": "Did you check performance the way this course has since NB09 -- a real holdout, not a number that flatters the model?",
    "Honest interpretation": "Did you say what the result actually means, and where it might be wrong -- the same discipline every NB14-NB20 case study has practiced?",
    "Justified choices": "Can you explain *why* you picked this model over an alternative, not just that it scored well?",
    "Clarity under questions": "Can you defend a choice you're asked about on the spot, not just read from a prepared script?",
}

for criterion, question in rubric.items():
    print(f"- {criterion}\n    {question}\n")
```

**None of these five criteria require your model to be the best-performing one in the class.** A modest, honestly-evaluated, well-explained result defends far better than an unexplainable high score — exactly the standard this course's own notebooks have held themselves to throughout (`NB04`'s mooring-line result, `NB16`'s honest French-classification limitations, `NB20`'s survey-density caveats — none of them hid a weak or surprising result, they explained it).

## 4. A defense template, drawn from this course's own case studies


Every real case study this course has taught, from `NB14` to `NB20`, has followed the same underlying structure without ever naming it explicitly. Naming it now gives you a genuine, reusable template for your own defense — not an arbitrary format, but the exact shape of "an honest engineering analysis," seen six times already.


```python
defense_template = [
    ("1. The real question", "What are you actually trying to predict or classify, and why does it matter for naval/ocean engineering?"),
    ("2. The real data", "Where did it come from, what does it actually contain, and what did you have to clean or exclude (and why)?"),
    ("3. Exploration", "What did you learn about the data before modeling it -- what did a real plot or summary statistic tell you?"),
    ("4. Method and justification", "What model(s) did you try, and why -- tied back to what NB07-NB17 taught about when each technique fits?"),
    ("5. Honest evaluation", "How did you test it -- a real, appropriate holdout -- and what did the result actually show, unembellished?"),
    ("6. Limitations", "What would you flag as a caveat if a colleague relied on this result? What's the weakest part of your analysis?"),
]

for section, prompt in defense_template:
    print(f"{section}\n   {prompt}\n")
```

This is deliberately the same six-part shape as this course's own case-study notebooks' section structure — you are not inventing a new format, you are applying one you've already seen in action six times.

## 5. Worked example: defending `NB20` with the template


`NB20` is fair game for this worked example precisely because you've already seen it in full — nothing here previews your individual project. The real, already-confirmed results from that class (real GMRT bathymetry, real kNN survey-density experiment) slot directly into the template above.


```python
nb20_defense_example = {
    "1. The real question": "Given only a sparse set of real depth soundings, how well can seafloor depth be predicted in between them? -- a genuine hydrographic-survey cost/accuracy trade-off.",
    "2. The real data": "Real bathymetry for the Cartagena coast/shelf, downloaded live and with no login from GMRT's public GridServer API -- a synthesis of real multibeam surveys and satellite-derived estimates.",
    "3. Exploration": "42.5% land / 57.5% sea in this bounding box; real depths down to ~2,490 m; visualized on a real cartopy map before any modeling.",
    "4. Method and justification": "kNN regression, chosen specifically because its 'nearby points are similar' assumption matches spatial interpolation -- compared honestly against Random Forest and a naive nearest-sounding baseline, not assumed to be best.",
    "5. Honest evaluation": "A genuine held-out set of real depth values, evaluated at multiple real survey densities: RMSE fell from 127.37 m at 0.2% coverage to 16.44 m at 10% coverage -- a real, monotonic, diminishing-returns relationship.",
    "6. Limitations": "GMRT's own 'ground truth' blends real soundings with satellite-estimated bathymetry in gaps; random-point sparsity is an easier case than a real line survey; sparse interpolation can't recover features smaller than the survey's own spacing.",
}

for section, content in nb20_defense_example.items():
    print(f"{section}\n   {content}\n")
```

Notice what's absent from this worked example: no claim that kNN is "the best" model in any universal sense, no hidden caveat, no result presented without its real limitation. **That is what a strong defense sounds like** — not a flawless result, an honestly explained one.
