## 3. La rúbrica de evaluación, en profundidad


La guía docente indica que el proyecto individual se evalúa por "resolución individual y exposición" — resolución *y* exposición, ambas obligatorias. En la práctica ya establecida de este curso (cada caso de estudio desde `NB14`), las cualidades que realmente distinguen un resultado sólido de uno débil son consistentes, y rara vez se reducen a "la cifra final de accuracy":

```python
# La rúbrica de evaluación: {criterio: pregunta que deberías saber responder}
rubric = {
    "Comprensión real de los datos": "¿Sabes explicar qué significa cada columna y por qué elegiste tus variables, y no solo que cargaste un CSV?",
    "Una evaluación genuina y adecuada": "¿Comprobaste el rendimiento como se hace en este curso desde NB10 (un conjunto reservado real), y no con una cifra que favorece al modelo?",
    "Interpretación honesta": "¿Dijiste qué significa realmente el resultado y dónde podría fallar, con la misma disciplina que han practicado todos los casos de estudio de NB14 a NB20?",
    "Decisiones justificadas": "¿Sabes explicar *por qué* elegiste este modelo frente a una alternativa, y no solo que obtuvo buena puntuación?",
    "Claridad ante las preguntas": "¿Sabes defender en el momento una decisión por la que te preguntan, sin limitarte a leer un guion preparado?",
}

# Mostrar cada criterio con su pregunta sangrada debajo
for criterion, question in rubric.items():
    print(f"- {criterion}\n    {question}\n")
```

**Ninguno de estos cinco criterios exige que tu modelo sea el de mejor desempeño de la clase.** `Un resultado modesto, evaluado honestamente y bien explicado se defiende muchísimo mejor que una puntuación alta inexplicable` — exactamente el estándar al que se han sometido los propios notebooks de este curso en todo momento (el resultado del amarre de `NB04`, las limitaciones honestas de la clasificación francesa de `NB19`, las advertencias sobre densidad de levantamiento de `NB20` — ninguno de ellos escondió un resultado débil o sorprendente, lo explicaron).

**Pruébalo tú mismo**: antes incluso de empezar tu proyecto, rellena esta misma rúbrica como autocomprobación a la que podrás volver más adelante — ahora mismo es perfectamente válido dejar todas las respuestas en blanco.

```python
# Una respuesta vacía por cada criterio de la rúbrica: rellénalas a medida que avance tu proyecto
self_assessment = {criterion: "" for criterion in rubric}

# Una cadena vacía cuenta como False, así que `or` muestra el recordatorio hasta que escribas algo
for criterion in self_assessment:
    print(f"- {criterion}: {self_assessment[criterion] or '(todavía no: vuelve aquí después de tu primer checkpoint)'}")
```

## 4. Una plantilla de defensa, extraída de los propios casos de estudio del curso


Cada caso de estudio real que este curso ha impartido, de `NB14` a `NB20`, ha seguido la misma estructura subyacente sin nombrarla nunca de forma explícita. Nombrarla ahora te da una plantilla genuina y reutilizable para tu propia defensa — no un formato arbitrario, sino la forma exacta de "un análisis de ingeniería honesto", vista ya seis veces.

```python
# La plantilla de defensa en seis partes: (sección, pregunta guía)
defense_template = [
    ("1. La pregunta real", "¿Qué intentas predecir o clasificar realmente, y por qué importa para la ingeniería naval/oceánica?"),
    ("2. Los datos reales", "¿De dónde proceden, qué contienen realmente y qué tuviste que limpiar o excluir (y por qué)?"),
    ("3. Exploración", "¿Qué aprendiste de los datos antes de modelarlos? ¿Qué te dijo una gráfica o un estadístico resumen real?"),
    ("4. Método y justificación", "¿Qué modelo(s) probaste y por qué, en relación con lo que NB07-NB17 enseñaron sobre cuándo encaja cada técnica?"),
    ("5. Evaluación honesta", "¿Cómo lo pusiste a prueba (con un conjunto reservado real y adecuado) y qué mostró realmente el resultado, sin adornos?"),
    ("6. Limitaciones", "¿Qué advertencia harías si un colega se basara en este resultado? ¿Cuál es la parte más débil de tu análisis?"),
]

for section, prompt in defense_template:
    print(f"{section}\n   {prompt}\n")
```

Esto es deliberadamente la misma estructura de seis partes que la estructura de secciones de los propios notebooks de casos de estudio de este curso — no estás inventando un formato nuevo, estás aplicando uno que ya has visto en acción seis veces.

**Pruébalo tú mismo**: elige `NB18` o `NB19` (no `NB20`, que ya se trabaja más abajo) y rellena los seis puntos de la plantilla de defensa a partir de tu propio recuerdo de esa clase — el objetivo es practicar la *estructura*, no memorizar cifras exactas, así que es válido escribir `(buscarlo)` para cualquier cosa que no recuerdes.

```python
my_case_study = "NB18"  # cámbialo a "NB19" si prefieres practicar con ese

# Una respuesta vacía por cada sección de la plantilla
my_defense_draft = {section: "" for section, _ in defense_template}

# Mostrar cada pregunta con tu respuesta, o un recordatorio mientras siga vacía
for section, prompt in defense_template:
    print(f"{section}\n   Pregunta: {prompt}\n   Tu respuesta: {my_defense_draft[section] or '(rellénalo)'}\n")
```
