### Variables y tipos básicos

Python tiene tipado dinámico: `el tipo de una variable se infiere del valor que se le asigna, y puede cambiar si se reasigna`.

| Tipo | Ejemplo | Ejemplo con sabor naval |

```python
ship_name = "MV Poseidon"
length_m = 209.9
num_engines = 2
is_operational = True

print(ship_name, length_m, num_engines, is_operational)
print(type(ship_name), type(length_m), type(num_engines), type(is_operational))
```

### Colecciones: listas, tuplas y diccionarios

- Una **lista** (`[...]`) es una secuencia ordenada y mutable.
- Una **tupla** (`(...)`) es como una lista pero inmutable (no se puede cambiar tras crearla).
- Un **dict** (`{...}`) asocia claves a valores — ideal para representar un registro (p. ej., un buque) con campos con nombre.

```python
fleet = ["MV Poseidon", "SS Neptune", "MV Triton", "SS Amphitrite"]
print(fleet[0])       # first element
print(fleet[-1])      # last element
print(fleet[1:3])     # slicing: elements 1 and 2
print(len(fleet))     # number of elements
```

Ahora un **dict**, para representar un buque como un único registro con campos con nombre:

```python
ship = {
    "name": "MV Poseidon",
    "length_m": 209.9,
    "displacement_t": 51252.5,
    "type": "Cargo Ship",
}

print(ship["name"], "-", ship["type"])
ship["is_operational"] = True  # dicts are mutable: we can add new keys
print(ship)
```

### Control de flujo: `if`, `for`, `while`

Python usa la **indentación** (no llaves) para marcar los bloques de código.

```python
for name, length in zip(fleet, [209.9, 320.0, 45.3, 180.2]):
    if length > 200:
        print(f"{name}: {length} m -> large vessel")
    elif length > 100:
        print(f"{name}: {length} m -> medium vessel")
    else:
        print(f"{name}: {length} m -> small vessel")
```

### Funciones

Las funciones empaquetan lógica reutilizable, toman parámetros, y (normalmente) `return`-an un valor.

```python
def classify_by_displacement(displacement_t):
    """Classify a vessel by its displacement in tonnes."""
    if displacement_t > 50000:
        return "large"
    elif displacement_t > 10000:
        return "medium"
    else:
        return "small"

for d in [51252.5, 15000, 3000]:
    print(d, "->", classify_by_displacement(d))
```

**Pruébalo tú mismo**: escribe una función `classify_by_length(length_m)` que devuelva `"long"` si el buque mide más de 200 m, `"medium"` si mide más de 100 m, y `"short"` en el resto de casos — misma forma que `classify_by_displacement` de arriba. Pruébala con `[209.9, 320.0, 45.3, 180.2, 260.5]`, y después compara con la celda de abajo.

```python
ship_lengths = [209.9, 320.0, 45.3, 180.2, 260.5]

def classify_by_length(length_m):
    """Classify a vessel by its length in meters."""
    if length_m > 200:
        return "long"
    elif length_m > 100:
        return "medium"
    else:
        return "short"

for length in ship_lengths:
    print(length, "->", classify_by_length(length))
```

### Comprensiones de listas (list comprehensions)

Una forma compacta y muy habitual de construir una lista nueva a partir de otra existente, filtrándola opcionalmente.

> **Para saber más**: [tutorial de Python: list comprehensions](https://docs.python.org/3/tutorial/datastructures.html#list-comprehensions).

```python
lengths = [209.9, 320.0, 45.3, 180.2, 260.5]

# Classic loop
large_ships = []
for length in lengths:
    if length > 150:
        large_ships.append(length)

# Equivalent list comprehension
large_ships_comp = [length for length in lengths if length > 150]

print(large_ships)
print(large_ships_comp)
print(large_ships == large_ships_comp)
```

Vamos a usar `%run` con un pequeño script de Fibonacci:

```python
%%writefile fib.py

def fib(N):
    """
    Return a list of the first N Fibonacci numbers.
    """
    f0, f1 = 0, 1
    f = [1] * N
    for n in range(1, N):
        f[n] = f0 + f1
        f0, f1 = f1, f[n]

    return f

print(fib(10))
```

**Ejercicio**: crea una nueva versión de la secuencia de Fibonacci que incluya el 0. Llámala `fib2.py` y escríbela usando `%%writefile`, exactamente como arriba.

```python
%%writefile fib2.py

def fib(N):
    """
    Return a list of the first N Fibonacci numbers starting from 0.
    """
    if N <= 0:
        return []

    f0, f1 = 0, 1
    f = [0] * N
    if N > 1:
        f[1] = 1

    for n in range(2, N):
        f[n] = f0 + f1
        f0, f1 = f1, f[n]

    return f
```

Para un ejemplo de perfilado más realista, define una pequeña simulación con NumPy: N buques aleatorios, cada uno con un paseo aleatorio de M pasos:

```python
import numpy as np

def random_ship_max_distance(M, N):
    """
    Simulate N random ships, each taking a random walk of M steps, and return the largest distance
    from the starting point achieved by any of the random walkers.
    """
    trajectories = [np.random.randn(M).cumsum() for _ in range(N)]
    return np.max(np.abs(trajectories))
```

Perfílala con `%prun` para ver dónde se va realmente el tiempo:

```python
%prun random_ship_max_distance(400, 10000)
```

La salida del perfilador desglosa cuántas veces se llamó a cada función y a dónde fue el tiempo acumulado. Mira tu propia salida: ¿es el cuello de botella el bucle a nivel de Python sobre los `N` buques, o las propias operaciones con arrays? La respuesta exacta puede variar entre versiones de NumPy y máquinas — es exactamente el tipo de pregunta que `%prun` responde directamente, en vez de suponerla.

**Pruébalo tú mismo**: `random_ship_max_distance` de arriba usa una list comprehension de Python (una trayectoria cada vez). Escribe una versión completamente vectorizada — sin bucle de Python, usando `np.random.randn(N, M).cumsum(axis=1)` para calcular las `N` trayectorias a la vez — y usa `%timeit` para comparar ambas versiones. ¿Cuánto más rápida es la versión vectorizada?

```python
def random_ship_max_distance_vectorized(M, N):
    """Vectorized version: all N trajectories computed at once, no Python loop."""
    trajectories = np.random.randn(N, M).cumsum(axis=1)
    return np.max(np.abs(trajectories))

# Same seed before each call -> both versions draw the exact same random numbers,
# in the exact same order, so this actually proves they compute the same thing
np.random.seed(0)
result_loop = random_ship_max_distance(400, 10000)
np.random.seed(0)
result_vectorized = random_ship_max_distance_vectorized(400, 10000)
print(f"Loop version:       {result_loop:.6f}")
print(f"Vectorized version: {result_vectorized:.6f}")
print("Identical result:", result_loop == result_vectorized)

# Now benchmark both (unseeded, since %timeit runs each statement many times anyway)
%timeit random_ship_max_distance(400, 10000)
%timeit random_ship_max_distance_vectorized(400, 10000)
```

Mismo resultado, calculado de dos formas distintas — confirmado arriba fijando la misma semilla en ambas llamadas antes de compararlas (sin una semilla fija, cada llamada genera números aleatorios nuevos, así que las dos versiones *no* devolverían el mismo número por coincidencia). Aquí `axis=1` colapsa los pasos propios de cada trayectoria en una suma acumulada por buque, la misma idea de `axis` que `NB03` cubre en profundidad.

## Parte 3 — Fundamentos de NumPy para trabajar con datos

[NumPy](https://numpy.org/) es la columna vertebral numérica bajo Pandas, scikit-learn, y toda librería de deep learning que usaremos más adelante. Su objeto central es el `ndarray`: un array rápido, homogéneo y multidimensional (un repaso completo de formas, dtypes y operaciones matriciales queda fuera del alcance de hoy; aquí solo cubrimos lo que necesitamos para el resto del curso).

### Creación de arrays y atributos básicos

```python
import numpy as np

lengths_m = np.array([209.9, 197.2, 213.0, 230.5, 195.3])
print(lengths_m)
print("shape:", lengths_m.shape, "| ndim:", lengths_m.ndim, "| dtype:", lengths_m.dtype)
```

### Indexado, slicing, y máscaras booleanas

Los arrays se indexan como las listas de Python, pero además admiten **máscaras booleanas**: indexar con una condición devuelve solo los elementos donde esa condición es `True` — `un patrón que usarás constantemente con datasets reales`.

> **Para saber más**: [NumPy: indexado sobre ndarrays](https://numpy.org/doc/stable/user/basics.indexing.html).

```python
print(lengths_m[0])         # first element
print(lengths_m[1:3])       # slice: elements 1 and 2
print(lengths_m > 200)      # boolean mask
print(lengths_m[lengths_m > 200])  # only the "large" ships
```

### Operaciones vectorizadas y estadísticas básicas

Las operaciones de NumPy se aplican **elemento a elemento**, sin escribir un bucle explícito — `esto es a la vez más conciso y mucho más rápido que Python puro` (recuerda `%timeit` de la Parte 2).

```python
lengths_ft = lengths_m * 3.28084  # meters to feet, applied to every element at once
print(lengths_ft.round(1))

print("mean:", lengths_m.mean().round(1), "m")
print("std:", lengths_m.std().round(1), "m")
print("min/max:", lengths_m.min(), "/", lengths_m.max(), "m")
```

### Arrays 2D: filas como buques, columnas como características

Un array 2D es el equivalente en NumPy de una pequeña tabla — conceptualmente la misma forma que usará un DataFrame de Pandas en la Parte 4, solo que sin nombres de columna.

```python
# columns: [length_m, beam_m, draft_m]
fleet_specs = np.array([
    [209.9, 33.0, 9.9],
    [197.2, 30.8, 11.4],
    [213.0, 30.0, 11.5],
])

print(fleet_specs.shape)   # (3 ships, 3 features)
print(fleet_specs[:, 0])   # all lengths (first column)
print(fleet_specs.mean(axis=0))  # mean of each column (feature)
```

**Pruébalo tú mismo**: usando `fleet_specs` y una máscara booleana, selecciona solo los buques con `beam_m` (columna 1) mayor que 30 — ¿cuántos hay, y cuál es el más largo de ellos? (Pista: combina una máscara booleana sobre la columna 1 con `.max()` sobre la columna 0.)

```python
wide_ships = fleet_specs[fleet_specs[:, 1] > 30]
print(f"{len(wide_ships)} ships with beam > 30 m")
print("Longest among them:", wide_ships[:, 0].max(), "m")
```

`.corr()` calcula la correlación por pares entre columnas numéricas (de -1 a 1) — `una forma rápida de detectar qué medidas se mueven juntas`, antes de decidir qué darle de comer a un modelo más adelante en el curso.

> **Para saber más**: [coeficiente de correlación de Pearson (Wikipedia)](https://en.wikipedia.org/wiki/Pearson_correlation_coefficient) · [documentación de `pandas.DataFrame.corr`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.corr.html).

```python
ships[["Length_m", "Beam_m", "Draft_m", "Displacement_t"]].corr().round(2)
```

**Pruébalo tú mismo**: combina filtrado y agrupación — entre los buques de más de 150 m, ¿cuál es el desplazamiento medio por tipo de buque?

```python
long_ships = ships[ships["Length_m"] > 150]
long_ships.groupby("Ship_Type")["Displacement_t"].mean().round(1)
```

## Gráficos exploratorios

Vamos a visualizar longitud frente a desplazamiento, coloreado por tipo de buque, y la distribución de una variable.

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
for ship_type, group in ships.groupby("Ship_Type"):
    ax.scatter(group["Length_m"], group["Displacement_t"], label=ship_type, alpha=0.7)

ax.set_xlabel("Length (m)")
ax.set_ylabel("Displacement (t)")
ax.set_title("Ship Length vs. Displacement by Type")
ax.legend()
ax.grid(True)
plt.show()
```

Y la distribución de una única variable:

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(ships["Displacement_t"], bins=15, edgecolor="black")
ax.set_xlabel("Displacement (t)")
ax.set_ylabel("Number of ships")
ax.set_title("Distribution of Ship Displacement")
plt.show()
```

> **Idea clave**: cargar → inspeccionar (`head`, `info`, `describe`) → filtrar/ordenar → agrupar → correlacionar → visualizar. Repetirás exactamente este flujo de EDA con cada dataset nuevo durante el resto del curso, antes de entrenar ningún modelo.
