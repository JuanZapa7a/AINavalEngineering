## Parte 1 — Historia de la Inteligencia Artificial *(≈45 min · Teoría)*

Artificial Intelligence (AI) has evolved significantly since its inception, transforming from theoretical concepts into practical applications that impact almost every aspect of modern life. [Artificial Intelligence (AI)](https://en.wikipedia.org/wiki/Artificial_intelligence) is a branch of [computer science](https://en.wikipedia.org/wiki/Computer_science) that focuses on creating systems capable of performing tasks that typically require human intelligence. These tasks include:

- [Problem-solving](https://en.wikipedia.org/wiki/Problem_solving)

- [Language understanding](https://en.wikipedia.org/wiki/Natural_language_processing)

### **Importancia de la IA en la ingeniería y otras disciplinas**

AI plays a fundamental role in various disciplines beyond computer science. In [medicine](https://en.wikipedia.org/wiki/Artificial_intelligence_in_healthcare), AI assists in disease diagnosis, medical imaging, and personalized treatments. ## **De 1950 a 1980: optimismo inicial y primeros logros vs. el "invierno de la IA"**

Although AI began with high expectations, its early years were marked by both significant achievements and limitations that revealed the complexity of replicating human intelligence.

### **Desarrollo de los primeros programas de IA**
1. **[Logic Theorist](https://en.wikipedia.org/wiki/Logic_Theorist?utm_source=chatgpt.com)**: Developed by Allen Newell and Herbert Simon, this program could prove mathematical theorems using logical rules. It marked a milestone by demonstrating machines could perform tasks previously considered exclusive to humans.
2. **[General Problem Solver (GPS)](https://en.wikipedia.org/wiki/General_Problem_Solver?utm_source=chatgpt.com)**: An attempt to create a universal program capable of solving any formalized problem. While successful in specific domains, its scope remained limited.
3. **[ELIZA](https://en.wikipedia.org/wiki/ELIZA?utm_source=chatgpt.com) and Human Language Simulation**:
   - Created by [Joseph Weizenbaum](https://en.wikipedia.org/wiki/Joseph_Weizenbaum?utm_source=chatgpt.com) in 1966, ELIZA was one of the first programs to simulate human conversation. Using simple pattern-matching rules, it could hold basic dialogues, particularly in simulated psychological therapy sessions.
   - Though ELIZA lacked true language comprehension, its ability to [deceive users](https://www.bbc.com/mundo/noticias-44290222?utm_source=chatgpt.com) into believing they were interacting with a human was groundbreaking.

### **Primeras aplicaciones de la IA**
- **Juegos como Damas y Ajedrez**: AI was applied early to strategy games. In 1948, Alan Turing developed an early chess program, manually simulated but foundational for future research in this [field](https://www.bigdata.uma.es/los-primeros-programas-de-ia-capaces-de-aprender/?utm_source=chatgpt.com).
- **Sistemas expertos como [DENDRAL](https://www.britannica.com/technology/DENDRAL?utm_source=chatgpt.com) y [MYCIN](https://www.britannica.com/technology/MYCIN?utm_source=chatgpt.com)**:
  - **DENDRAL** (1965) helped chemists interpret spectrometric data and hypothesize molecular structures.
  - **MYCIN** (1970) focused on diagnosing infectious blood diseases, showcasing AI's potential in [medical decision-making](https://blogthinkbig.com/sistemas-expertos-inteligencia-artificial?utm_source=chatgpt.com).

### **El "invierno de la IA"**
- **Desafíos**: los primeros sistemas de IA carecían de:
  1. Autonomous learning capabilities (due to inadequate algorithms).
  2. Sufficient computational power (limited hardware).
  3. Access to large datasets (databases were nonexistent).
- **El informe Lighthill (1973)**: Commissioned by the UK government, this report criticized AI's unfulfilled promises, particularly in areas like [machine translation](https://translatorstudio.co.uk/machine-translation-history/), and highlighted technological limitations. It led to drastic cuts in [UK AI research funding](https://iaresponde.com/invierno-de-la-ia/?utm_source=chatgpt.com).

### **Legado de la época**
The 1950s–1970s laid the groundwork for modern AI, proving machines could tackle logic, language, and specialized tasks. Innovations like ELIZA and expert systems paved the way for future advancements in machine learning and neural networks.

## **De 1980 a 2000: resurgimiento y consolidación**

### **El auge de los sistemas expertos**

The 1980s marked a pivotal era for artificial intelligence (AI) with the widespread adoption of **[expert systems](https://en.wikipedia.org/wiki/Expert_system)**—rule-based programs designed to emulate human expertise in specialized domains.

- En **medicina**, sistemas como **[MYCIN](https://en.wikipedia.org/wiki/Mycin)** y **[DXplain](https://en.wikipedia.org/wiki/DXplain)** demonstrated the potential of AI in diagnostics.
- En **ingeniería**, expert systems like **[XCON](https://en.wikipedia.org/wiki/XCON)** optimized manufacturing processes, while **[PROSPECTOR](https://en.wikipedia.org/wiki/Prospector_(expert_system))** aided geologists in mineral exploration with 85% accuracy.
- El **sector empresarial** embraced systems like **[CLIPS](https://en.wikipedia.org/wiki/CLIPS)** for fraud detection and financial forecasting. By the late 1980s, the global expert systems market was valued at over $1 billion.

Sin embargo, los sistemas expertos dependían de la **creación manual de reglas** (el llamado *knowledge acquisition problem*) y no podían aprender de datos nuevos.

### **Avances en algoritmos y hardware**

- En 1986, el **[algoritmo de backpropagation](https://en.wikipedia.org/wiki/Backpropagation)** — impulsado por investigadores como **[Geoffrey Hinton](https://en.wikipedia.org/wiki/Geoffrey_Hinton)** — revolucionó el entrenamiento de redes neuronales.
- En 1989, [Yann LeCun](https://scholar.google.com/citations?user=WLN3QrAAAAAJ&hl=es) demostró la arquitectura **[LeNet](https://en.wikipedia.org/wiki/LeNet)**, una red neuronal convolucional (CNN) capaz de reconocer dígitos manuscritos.
- El auge del **[procesamiento en paralelo](https://en.wikipedia.org/wiki/Parallel_computing)** y las **[GPU](https://en.wikipedia.org/wiki/Graphics_processing_unit)** aceleró los cálculos, sentando las bases de los avances del siglo XXI.

### **Hitos destacados**

- En 1997, **[Deep Blue](https://en.wikipedia.org/wiki/Deep_Blue_(chess_computer))** de IBM derrotó al campeón mundial Garry Kasparov, evaluando 200 millones de posiciones por segundo.
- El proyecto **[ALVINN](https://en.wikipedia.org/wiki/ALVINN)** (1989, Carnegie Mellon) pionero en conducción autónoma mediante una red neuronal, precursor de compañías como Waymo y Tesla.

### **Legado de la época**

The 1980s–1990s transformed AI from an academic curiosity into a tool with real-world impact, sentando las bases de innovaciones como **[AlphaGo](https://en.wikipedia.org/wiki/AlphaGo)** y los modelos de lenguaje modernos.

## **El siglo XXI: la era del Big Data y el Deep Learning**

The 21st century has witnessed an unprecedented explosion of [data](https://en.wikipedia.org/wiki/Data) and [computational power](https://en.wikipedia.org/wiki/Computational_power), fueling advancements in AI. Esta era se define por dos componentes clave: **[big data](https://en.wikipedia.org/wiki/Big_data)** y **[deep learning](https://en.wikipedia.org/wiki/Deep_learning)**.

### **El papel del Big Data**

1. **[Improved Model Training](https://en.wikipedia.org/wiki/Training_data)**: la disponibilidad de grandes datasets mejora la precisión y generalización de los modelos.
2. **[Data-Driven Decision Making](https://en.wikipedia.org/wiki/Data-driven_decision_making)**: los modelos de IA analizan grandes volúmenes de datos para ayudar en la toma de decisiones.
3. **[Personalized Experiences](https://en.wikipedia.org/wiki/Personalization)**: desde sistemas de recomendación hasta publicidad dirigida.

### **Deep Learning: un enfoque revolucionario**

[Deep learning](https://en.wikipedia.org/wiki/Deep_learning), a subset of machine learning, has emerged as the dominant approach in AI research and applications. Unlike traditional machine learning techniques, deep learning models automatically extract features from raw data, eliminando la necesidad de una extensa ingeniería manual de features.

**Pioneros del Deep Learning**: [Geoffrey Hinton](https://en.wikipedia.org/wiki/Geoffrey_Hinton) (backpropagation, deep belief networks), [Yann LeCun](https://en.wikipedia.org/wiki/Yann_LeCun) (redes neuronales convolucionales — CNN), y [Yoshua Bengio](https://en.wikipedia.org/wiki/Yoshua_Bengio) (modelado de secuencias).

### **Conceptos básicos de IA**

Artificial intelligence (AI) is a field of computer science focused on developing systems capable of performing tasks that typically require human intelligence, such as learning, reasoning, problem-solving, perception, and decision-making. It encompasses various subfields: machine learning, deep learning, natural language processing, computer vision, and robotics.

1. **[Machine Learning (ML)](https://en.wikipedia.org/wiki/Machine_learning)**: a core subset of AI where systems learn from data without being explicitly programmed. Categorized into supervised, unsupervised, semi-supervised and reinforcement learning.

5. **[Generative AI](https://en.wikipedia.org/wiki/Generative_artificial_intelligence)**: creates new, original content (texto, imágenes, audio, código) identificando patrones en grandes datasets.

   *Ejemplo:* [revolucionando el diseño de buques con IA y Machine Learning](https://www.bmt.org/insights/revolutionising-ship-design-with-artificial-intelligence-and-machine-learning/).

### **Tipos de IA: IA débil vs. IA fuerte**

| Característica | **IA débil (Narrow AI)** | **IA fuerte (AGI)** |
|----------------------|--------------------------------|------------------------------------|
| Alcance | Estrecho, específico de una tarea | Amplio, propósito general |
| Consciencia | No | Teórica |
| Razonamiento a nivel humano | No | Sí (en teoría) |
| Ejemplos | Siri, filtros de spam, sistemas de recomendación | HAL 9000, Data (*ficticios* — no existe ningún sistema real hoy) |
| Aprendizaje y adaptación | Limitado a su dominio | Flexible y auto-mejorable |

| Tipo | Descripción | Perspectiva marítima |
|------|-------------|----------------------|
| IA débil (Narrow AI) | Especializada en una sola tarea. No generaliza más allá de ella. | Piloto automático, asistente de navegación. |
| IA fuerte (General AI) | Inteligencia a nivel humano, flexible y adaptable. | Hipotética: un buque totalmente autónomo capaz de decidir como un capitán en cualquier escenario. |

> **Idea clave**: casi toda la IA que usamos hoy —incluida toda la de este curso— es **IA débil**. La IA fuerte (AGI) sigue siendo un objetivo de investigación, no una realidad.

### **Ejemplo práctico: predicción de fallos por vibraciones de motor** *(≈15 min · Práctica)*

Vamos a simular un caso muy simplificado de mantenimiento predictivo: un sensor de vibración de un motor de buque que registra datos cada hora durante un día.


```python
# Visualicemos un caso de uso
import matplotlib.pyplot as plt
import numpy as np

# Datos de vibración del motor simulados
np.random.seed(0)
hours = np.arange(0, 24, 1)
vibration = 0.3 * np.sin(0.3 * hours) + np.random.normal(0, 0.1, size=hours.shape)

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, marker='o')
plt.title("Vibración simulada del motor durante 24 horas")
plt.xlabel("Tiempo (horas)")
plt.ylabel("Amplitud de vibración")
plt.grid(True)
plt.show()
```

Si etiquetamos cuándo la vibración supera un umbral, la IA puede aprender a predecir fallos con antelación.

```python
thresh = 0.25
alerts = vibration > thresh

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, label="Vibración")
plt.axhline(thresh, color='red', linestyle='--', label="Umbral")
plt.scatter(hours[alerts], vibration[alerts], color='red', label="Alertas")
plt.title("Umbral de mantenimiento predictivo")
plt.xlabel("Tiempo (horas)")
plt.ylabel("Amplitud de vibración")
plt.legend()
plt.grid(True)
plt.show()
```
