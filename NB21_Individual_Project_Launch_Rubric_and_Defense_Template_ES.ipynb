## 3. La rúbrica de evaluación, en profundidad


La guía docente indica que el proyecto individual se evalúa por "resolución individual y exposición" — resolución *y* exposición, ambas obligatorias. En la práctica ya establecida de este curso (cada caso de estudio desde `NB14`), las cualidades que realmente distinguen un resultado sólido de uno débil son consistentes, y rara vez se reducen a "la cifra final de accuracy":

```python
rubric = {
    "Real understanding of the data": "Can you explain what every column means, and why you chose your features -- not just that you loaded a CSV?",
    "A genuine, appropriate evaluation": "Did you check performance the way this course has since NB10 -- a real holdout, not a number that flatters the model?",
    "Honest interpretation": "Did you say what the result actually means, and where it might be wrong -- the same discipline every NB14-NB20 case study has practiced?",
    "Justified choices": "Can you explain *why* you picked this model over an alternative, not just that it scored well?",
    "Clarity under questions": "Can you defend a choice you're asked about on the spot, not just read from a prepared script?",
}

for criterion, question in rubric.items():
    print(f"- {criterion}\n    {question}\n")
```

**Ninguno de estos cinco criterios exige que tu modelo sea el de mejor desempeño de la clase.** `Un resultado modesto, evaluado honestamente y bien explicado se defiende muchísimo mejor que una puntuación alta inexplicable` — exactamente el estándar al que se han sometido los propios notebooks de este curso en todo momento (el resultado del amarre de `NB04`, las limitaciones honestas de la clasificación francesa de `NB19`, las advertencias sobre densidad de levantamiento de `NB20` — ninguno de ellos escondió un resultado débil o sorprendente, lo explicaron).

**Pruébalo tú mismo**: antes incluso de empezar tu proyecto, rellena esta misma rúbrica como autocomprobación a la que podrás volver más adelante — ahora mismo es perfectamente válido dejar todas las respuestas en blanco.

```python
self_assessment = {criterion: "" for criterion in rubric}

for criterion in self_assessment:
    print(f"- {criterion}: {self_assessment[criterion] or '(not yet -- revisit this after your first checkpoint)'}")
```

## 4. Una plantilla de defensa, extraída de los propios casos de estudio del curso


Cada caso de estudio real que este curso ha impartido, de `NB14` a `NB20`, ha seguido la misma estructura subyacente sin nombrarla nunca de forma explícita. Nombrarla ahora te da una plantilla genuina y reutilizable para tu propia defensa — no un formato arbitrario, sino la forma exacta de "un análisis de ingeniería honesto", vista ya seis veces.

```python
defense_template = [
    ("1. The real question", "What are you actually trying to predict or classify, and why does it matter for naval/ocean engineering?"),
    ("2. The real data", "Where did it come from, what does it actually contain, and what did you have to clean or exclude (and why)?"),
    ("3. Exploration", "What did you learn about the data before modeling it -- what did a real plot or summary statistic tell you?"),
    ("4. Method and justification", "What model(s) did you try, and why -- tied back to what NB07-NB17 taught about when each technique fits?"),
    ("5. Honest evaluation", "How did you test it -- a real, appropriate holdout -- and what did the result actually show, unembellished?"),
    ("6. Limitations", "What would you flag as a caveat if a colleague relied on this result? What's the weakest part of your analysis?"),
]

for section, prompt in defense_template:
    print(f"{section}\n   {prompt}\n")
```

Esto es deliberadamente la misma estructura de seis partes que la estructura de secciones de los propios notebooks de casos de estudio de este curso — no estás inventando un formato nuevo, estás aplicando uno que ya has visto en acción seis veces.

**Pruébalo tú mismo**: elige `NB18` o `NB19` (no `NB20`, que ya se trabaja más abajo) y rellena los seis puntos de la plantilla de defensa a partir de tu propio recuerdo de esa clase — el objetivo es practicar la *estructura*, no memorizar cifras exactas, así que es válido escribir `(buscarlo)` para cualquier cosa que no recuerdes.

```python
my_case_study = "NB18"  # change to "NB19" if you'd rather practice with that one

my_defense_draft = {section: "" for section, _ in defense_template}

for section, prompt in defense_template:
    print(f"{section}\n   Prompt: {prompt}\n   Your answer: {my_defense_draft[section] or '(fill this in)'}\n")
```

## 5. Ejemplo resuelto: defender `NB20` con la plantilla


`NB20` es un ejemplo legítimo para este caso resuelto precisamente porque ya lo has visto completo — nada de esto anticipa tu proyecto individual. Los resultados reales y ya confirmados de esa clase (la batimetría real de GMRT, el experimento real de densidad de levantamiento con kNN) encajan directamente en la plantilla de arriba.

```python
nb20_defense_example = {
    "1. The real question": "Given only a sparse set of real depth soundings, how well can seafloor depth be predicted in between them? -- a genuine hydrographic-survey cost/accuracy trade-off.",
    "2. The real data": "Real bathymetry for the Cartagena coast/shelf, downloaded live and with no login from GMRT's public GridServer API -- a synthesis of real multibeam surveys and satellite-derived estimates.",
    "3. Exploration": "42.5% land / 57.5% sea in this bounding box; real depths down to ~2,490 m; visualized on a real cartopy map before any modeling.",
    "4. Method and justification": "kNN regression, chosen specifically because its 'nearby points are similar' assumption matches spatial interpolation -- compared honestly against Random Forest and a naive nearest-sounding baseline, not assumed to be best.",
    "5. Honest evaluation": "A genuine held-out set of real depth values, evaluated at multiple real survey densities: RMSE fell from 127.37 m at 0.2% coverage to 16.44 m at 10% coverage -- a real, monotonic, diminishing-returns relationship.",
    "6. Limitations": "GMRT's own 'ground truth' blends real soundings with satellite-estimated bathymetry in gaps; random-point sparsity is an easier case than a real line survey; sparse interpolation can't recover features smaller than the survey's own spacing.",
}

for section, content in nb20_defense_example.items():
    print(f"{section}\n   {content}\n")
```

Fíjate en lo que está ausente de este ejemplo resuelto: `ninguna afirmación de que kNN sea "el mejor" modelo en ningún sentido universal, ninguna advertencia escondida, ningún resultado presentado sin su limitación real`. **Así es como suena una defensa sólida** — no un resultado impecable, sino uno explicado con honestidad.
