### What is Artificial Intelligence (AI)?

#### **Definition**
Artificial Intelligence (AI) is the field of computer science that focuses on creating systems capable of performing tasks that normally require human intelligence. These tasks include:

- Problem solving  
- Learning from data  
- Understanding natural language  
- Perception (e.g., vision or sound)  
- Decision-making  

> **Basic definition**: *AI is the simulation of human intelligence by machines.*

---

#### **Brief History of AI**

| Year | Milestone |
|------|-----------|
| **1950** | Alan Turing proposes the "Turing Test" to evaluate machine intelligence. |
| **1956** | The term *Artificial Intelligence* is coined at the Dartmouth Conference. |
| **1960s–70s** | Early AI programs solve math problems and play games like chess. |
| **1980s** | Expert systems become popular (e.g., MYCIN for medical diagnosis). |
| **1997** | IBM's Deep Blue defeats chess world champion Garry Kasparov. |
| **2012** | Deep Learning revolution begins with breakthroughs in image recognition. |
| **Today** | AI powers systems in robotics, autonomous ships, smart assistants, and more.

---

#### **Main Areas of AI**

| Area | Description | Naval Example |
|------|-------------|---------------|
| **Machine Learning (ML)** | Algorithms learn patterns from data. | Predict maintenance needs for ship engines. |
| **Computer Vision** | Interpreting images or video. | Identifying objects (e.g., vessels, ports) from satellite images. |
| **Natural Language Processing (NLP)** | Understanding and generating human language. | Chatbots for marine traffic communication. |
| **Robotics** | AI-powered physical machines. | Underwater drones for hull inspection. |
| **Expert Systems** | Rule-based decision-making. | Damage assessment systems in ship emergencies. |

---

#### **Why is AI relevant in Naval Engineering and Marine Science?**

- **Autonomous navigation** (surface & underwater)
- **Fuel optimization** using weather and route prediction
- **Early fault detection** in propulsion systems
- **Marine ecosystem monitoring** (using sensors & data analytics)
- **Port traffic control** and smart logistics

```python
# ## Let's Visualize a Use Case

# %% 
import matplotlib.pyplot as plt
import numpy as np

# Simulated engine vibration data
np.random.seed(0)
hours = np.arange(0, 24, 1)
vibration = 0.3 * np.sin(0.3 * hours) + np.random.normal(0, 0.1, size=hours.shape)

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, marker='o')
plt.title("Simulated Engine Vibration Over 24 Hours")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.grid(True)
plt.show()

# %% [markdown]
# ### 🔧 Can AI help?
# Yes! If we label when vibration goes above a threshold, AI can learn to predict failures in advance.

# %%
thresh = 0.25
alerts = vibration > thresh

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, label="Vibration")
plt.axhline(thresh, color='red', linestyle='--', label="Threshold")
plt.scatter(hours[alerts], vibration[alerts], color='red', label="Alerts")
plt.title("Predictive Maintenance Threshold")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```
