```python
import numpy as np
import pandas as pd

# Fijar la semilla para reproducibilidad
np.random.seed(42)

# Número de muestras
n = 768

# Generar variables con distribuciones normales
num_inspecciones = np.random.normal(3.8, 3.4, n).clip(0).round()
salinidad = np.random.normal(120.9, 32.0, n).clip(0)
presion_agua = np.random.normal(69.1, 19.4, n).clip(0)
espesor_recubrimiento = np.random.normal(20.5, 16.0, n).clip(0)
nivel_corrosion = np.random.normal(79.8, 115.2, n).clip(0, 100)
indice_mantenimiento = np.random.normal(0.5, 0.3, n).clip(0, 1)
edad_barco = np.random.normal(33.2, 11.8, n).clip(0)
funcion_riesgo = np.random.normal(0.4719, 0.331, n).clip(0)

# Generar variable binaria de resultado basada en una condición ficticia
resultado = (nivel_corrosion > 50).astype(int)

# Crear DataFrame
df = pd.DataFrame({
  'Num_Inspecciones': num_inspecciones,
  'Salinidad': salinidad,
  'Presion_Agua': presion_agua,
  'Espesor_Recubrimiento': espesor_recubrimiento,
  'Nivel_Corrosion': nivel_corrosion,
  'Indice_Mantenimiento': indice_mantenimiento,
  'Edad_Barco': edad_barco,
  'Funcion_Riesgo': funcion_riesgo,
  'Resultado': resultado
})

# Mostrar las primeras filas del DataFrame
print(df.head())

# Guardar el dataset en un archivo CSV
df.to_csv('corrosion_barcos_sintetico.csv', index=False)
```
