## Parte 1 — Historia de la Inteligencia Artificial

### **Definición**

La Inteligencia Artificial (IA) ha evolucionado de forma significativa desde sus inicios, pasando de conceptos teóricos a aplicaciones prácticas que afectan a casi todos los aspectos de la vida moderna. La [Inteligencia Artificial (IA)](https://en.wikipedia.org/wiki/Artificial_intelligence) es una rama de la [informática](https://en.wikipedia.org/wiki/Computer_science) centrada en `crear sistemas capaces de realizar tareas que normalmente requieren inteligencia humana`. Estas tareas incluyen:

- [Resolución de problemas](https://en.wikipedia.org/wiki/Problem_solving)
- [Aprendizaje a partir de datos](https://en.wikipedia.org/wiki/Machine_learning)
- [Comprensión del lenguaje](https://en.wikipedia.org/wiki/Natural_language_processing)
- [Percepción](https://en.wikipedia.org/wiki/Perception) (p. ej., visión o sonido)
- [Toma de decisiones](https://trailhead.salesforce.com/es/content/learn/modules/critical-thinking-and-decision-making-with-data-and-ai/discover-decision-making-frameworks-ai-models)

### **Importancia de la IA en la ingeniería y otras disciplinas**

La IA juega un papel fundamental en muchas disciplinas más allá de la informática. En [ingeniería](https://en.wikipedia.org/wiki/Engineering), la IA mejora la [automatización](https://en.wikipedia.org/wiki/Automation), el [mantenimiento predictivo](https://en.wikipedia.org/wiki/Predictive_maintenance) y la [robótica](https://en.wikipedia.org/wiki/Robotics), optimizando procesos industriales y mejorando la eficiencia. En [medicina](https://en.wikipedia.org/wiki/Artificial_intelligence_in_healthcare), la IA ayuda en el diagnóstico de enfermedades, la imagen médica y los tratamientos personalizados. También está transformando las [finanzas](https://en.wikipedia.org/wiki/Artificial_intelligence_in_finance) mediante el trading algorítmico y la evaluación de riesgos, mientras que en [educación](https://en.wikipedia.org/wiki/Artificial_intelligence_in_education) se usa para el aprendizaje personalizado y los [sistemas automáticos de calificación](https://dl.acm.org/doi/10.1145/3636515).

### **Breve historia de la IA**

**[Hitos clave en la historia de la IA](https://en.wikipedia.org/wiki/Timeline_of_artificial_intelligence)**

La IA ha pasado por varias fases significativas, cada una marcada por avances cruciales. Las fases siguientes se etiquetan por década ("años 50", "años 80", etc.) en vez de por año exacto — una etiqueta de década siempre significa el *tramo completo* de diez años ("años 50" = de 1950 a 1959), y "de los años 50 a los 70" significa todos los años desde 1950 hasta el final de los 70 (1979), no solo los dos años concretos 1950 y 1970. Leído así, las tres fases siguientes son contiguas — cada una empieza justo donde termina la década de la anterior, sin saltarse ningún año ni solaparse:

- **[Fundamentos y el primer invierno de la IA (años 50 a 70, es decir 1950–1979)](https://en.wikipedia.org/wiki/History_of_artificial_intelligence)**: fundamentos teóricos sentados por pioneros como [Alan Turing](https://en.wikipedia.org/wiki/Alan_Turing), los primeros programas de IA, y un primer [invierno de la IA](https://en.wikipedia.org/wiki/AI_winter) con recorte de financiación tras expectativas incumplidas.
- **[Sistemas expertos, resurgimiento y un segundo invierno de la IA (años 80 y 90, es decir 1980–1999)](https://en.wikipedia.org/wiki/Expert_system)**: los sistemas expertos basados en reglas llevaron la IA a la industria, seguidos de un segundo invierno de la IA cuando colapsó el mercado de hardware especializado que los sostenía.
- **[Big data y deep learning (años 2000–presente, es decir desde 2000)](https://en.wikipedia.org/wiki/Deep_learning)**: el paso de la IA basada en reglas al [machine learning](https://en.wikipedia.org/wiki/Machine_learning) basado en datos, y los avances en redes neuronales que impulsan el [procesamiento del lenguaje natural](https://en.wikipedia.org/wiki/Natural_language_processing), la [visión por computador](https://en.wikipedia.org/wiki/Computer_vision) y la [automatización impulsada por IA](https://www.rpatech.ai/ai-powered-automation/).

Al repasar estos hitos podemos entender mejor cómo la IA ha pasado de ser una idea abstracta a una tecnología esencial que sigue moldeando el futuro.

| Año | Hito |
|------|-----------|
| **1950** | Alan Turing propone el "[test de Turing](https://en.wikipedia.org/wiki/Turing_test)" para evaluar la inteligencia de una máquina. |
| **1956** | Se acuña el término *Inteligencia Artificial* en la [Conferencia de Dartmouth](https://self-transcendence.org/ai-history-the-dartmouth-conference). |
| **Años 60–70** | Los primeros programas de IA resuelven problemas matemáticos y juegan a juegos como el [ajedrez](https://computerhistory.org/blog/ai-and-play-part-1-how-games-have-driven-two-schools-of-ai-research/). |
| **Años 80** | Los [sistemas expertos](https://www.techtarget.com/searchenterpriseai/definition/expert-system) — construidos sobre prototipos de investigación de los 70 como MYCIN y DENDRAL — se adoptan ampliamente en la industria. |
| **1997** | El [Deep Blue de IBM](https://www.ibm.com/history/deep-blue) vence al campeón mundial de ajedrez Garry Kasparov. |
| **2012** | Comienza la revolución del deep learning con avances en [reconocimiento de imágenes](https://www.researchgate.net/publication/385379683_Deep_Learning_in_Image_Recognition_A_New_Era_for_Computer_Vision). |
| **Hoy** | La IA impulsa sistemas de robótica, buques autónomos, [asistentes inteligentes](https://aisera.com/chatbots-virtual-assistants-conversational-ai/) y mucho más. |

## **De los años 50 a los 70 (1950–1979): fundamentos y el primer invierno de la IA**

Aunque la IA empezó con expectativas muy altas, sus primeros años estuvieron marcados tanto por logros significativos como por limitaciones que revelaron `la complejidad de replicar la inteligencia humana`.

### **Desarrollo de los primeros programas de IA**
1. **[Logic Theorist](https://en.wikipedia.org/wiki/Logic_Theorist)**: desarrollado por Allen Newell y Herbert Simon, este programa podía demostrar teoremas matemáticos usando reglas lógicas. Marcó un hito al demostrar que las máquinas podían realizar tareas antes consideradas exclusivas de los humanos.
2. **[General Problem Solver (GPS)](https://en.wikipedia.org/wiki/General_Problem_Solver)**: un intento de crear un programa universal capaz de resolver cualquier problema formalizado. Aunque tuvo éxito en dominios concretos, su alcance siguió siendo limitado.
3. **[ELIZA](https://en.wikipedia.org/wiki/ELIZA) y la simulación del lenguaje humano**:
   - Creado por [Joseph Weizenbaum](https://en.wikipedia.org/wiki/Joseph_Weizenbaum) en 1966, ELIZA fue uno de los primeros programas en simular una conversación humana. Con reglas simples de coincidencia de patrones, podía mantener diálogos básicos, sobre todo en sesiones simuladas de terapia psicológica.
   - Aunque ELIZA no comprendía realmente el lenguaje, su capacidad para [engañar a los usuarios](https://www.bbc.com/mundo/noticias-44290222) haciéndoles creer que hablaban con una persona fue revolucionaria.

### **Primeras aplicaciones de la IA**
- **Juegos como las damas y el ajedrez**: la IA se aplicó pronto a juegos de estrategia. En 1948, Alan Turing desarrolló un primer programa de ajedrez, simulado a mano pero fundacional para la investigación posterior en este [campo](https://www.bigdata.uma.es/los-primeros-programas-de-ia-capaces-de-aprender/).
- **Sistemas expertos como [DENDRAL](https://www.britannica.com/technology/DENDRAL) y [MYCIN](https://www.britannica.com/technology/MYCIN)**:
  - **DENDRAL** (iniciado en 1964, en la Universidad de Stanford) ayudaba a los químicos a interpretar datos espectrométricos y a plantear hipótesis sobre estructuras moleculares.
  - **MYCIN** (1970) se centraba en el diagnóstico de enfermedades infecciosas de la sangre, mostrando el potencial de la IA en la [toma de decisiones médicas](https://blogthinkbig.com/sistemas-expertos-inteligencia-artificial). [Ver a MYCIN en acción](https://youtu.be/16Z5hNmfdxQ).

### **El "invierno de la IA"**
- **Retos**: los primeros sistemas de IA carecían de:
  1. Capacidad de aprendizaje autónomo (por algoritmos inadecuados).
  2. Suficiente potencia de cálculo (hardware limitado).
  3. Acceso a grandes conjuntos de datos (las bases de datos prácticamente no existían).
- **El informe Lighthill (1973)**: encargado por el gobierno británico, este informe criticó las promesas incumplidas de la IA, especialmente en áreas como la [traducción automática](https://translatorstudio.co.uk/machine-translation-history/), y puso de relieve sus limitaciones tecnológicas. Provocó recortes drásticos en la financiación de la investigación de IA en el Reino Unido. [Ver el debate Lighthill de 1973](https://youtu.be/03p2CADwGF8).
- **Impacto del [primer invierno de la IA](https://en.wikipedia.org/wiki/AI_winter) (aproximadamente de 1974 al cambio de década)**: el interés y la financiación de la IA se desplomaron a ambos lados del Atlántico. Sus efectos se prolongaron justo hasta donde arranca la siguiente sección: la investigación temprana en sistemas expertos (PROSPECTOR, XCON) ya estaba en marcha antes de que el invierno hubiera terminado del todo, y a principios de los 80 ese trabajo se convirtió visiblemente en el resurgimiento industrial que se trata a continuación. Estos hechos subrayan la naturaleza cíclica del progreso de la IA, donde los avances suelen ir seguidos de escepticismo y reevaluación — un *segundo* invierno de la IA llegaría a finales de los años 80 (ver la siguiente sección).

### **Legado de la época**
Los años 50 a 70 sentaron las bases de la IA moderna, demostrando que `las máquinas podían abordar la lógica, el lenguaje y tareas especializadas`. Aunque el optimismo se moderó con la realidad, innovaciones como ELIZA y los sistemas expertos abrieron camino a los avances posteriores en machine learning y redes neuronales. Las lecciones del invierno de la IA siguen siendo relevantes, y recuerdan a los investigadores que deben equilibrar la ambición con la viabilidad técnica.

## **Los años 80 y 90 (1980–1999): sistemas expertos, resurgimiento y un segundo invierno de la IA**

### **El auge de los sistemas expertos**

Los años 80 marcaron una época decisiva para la inteligencia artificial (IA) con la adopción generalizada de los **[sistemas expertos](https://en.wikipedia.org/wiki/Expert_system)** —programas basados en reglas diseñados para emular el conocimiento experto humano en dominios especializados—. Estos sistemas se convirtieron en la columna vertebral de las aplicaciones de IA en sectores como la **medicina, la ingeniería y los negocios**, revolucionando los procesos de toma de decisiones.

- En **medicina**, el sistema experto de los 70 **[MYCIN](https://en.wikipedia.org/wiki/Mycin)** (ver sección anterior) siguió marcando el campo: ***desarrollado en la Universidad de Stanford***, obtuvo una tasa de aceptabilidad del 65% por parte de especialistas en una evaluación de referencia — comparable a (y en ese estudio, superior a) las valoraciones del 42,5–62,5% de médicos en ejercicio. El sistema de los 80 **[DXplain](https://en.wikipedia.org/wiki/DXplain)** siguió esta idea, usando ***una base de conocimiento de enfermedades para sugerir diagnósticos a partir de síntomas***. Estas herramientas redujeron errores y agilizaron los flujos de trabajo en el ámbito sanitario.
- En **ingeniería**, sistemas expertos como **[XCON](https://en.wikipedia.org/wiki/XCON)** (desarrollado en 1978 por John McDermott en la Universidad Carnegie Mellon para Digital Equipment Corporation) ***optimizaba los procesos de fabricación configurando automáticamente los pedidos de ordenadores***. Por su parte, **[PROSPECTOR](https://www.sri.com/hoi/prospector-computer-based-expert-system/)** (SRI International) ***ayudaba a los geólogos en la exploración minera***, identificando con éxito un yacimiento real y hasta entonces desconocido de molibdeno en el estado de Washington. Estas aplicaciones subrayaron la capacidad de la IA para abordar problemas complejos y específicos de cada dominio.
- El **sector empresarial** adoptó sistemas como **[CLIPS](https://en.wikipedia.org/wiki/CLIPS)** para tareas que iban ***desde la detección de fraude hasta la previsión financiera***. Los bancos usaban la IA para evaluar el riesgo crediticio, mientras los minoristas la aprovechaban para la gestión de inventario. A finales de los 80, [dos tercios de las empresas del Fortune 500](https://en.wikipedia.org/wiki/Expert_system) habían adoptado sistemas expertos en su operativa diaria.

Sin embargo, pronto surgieron **limitaciones**. Los sistemas expertos dependían de la **creación manual de reglas**, lo que exigía una aportación humana muy costosa en tiempo — un cuello de botella conocido como el **[problema de adquisición del conocimiento](https://en.wikipedia.org/wiki/Knowledge_engineering)**. También tenían problemas de adaptabilidad; a diferencia de los humanos, ***no podían aprender de datos nuevos ni manejar situaciones ambiguas***. Los críticos argumentaban que estos sistemas eran mero "conocimiento congelado", incapaz de evolucionar. El [mercado de hardware de máquinas LISP](https://en.wikipedia.org/wiki/AI_winter) sobre el que corrían muchos sistemas expertos colapsó en 1987, y a principios de los 90 la mayoría de los proveedores especializados de sistemas expertos habían quebrado o reducido su actividad — un **segundo invierno de la IA**, distinto del de los años 70, aunque el legado de la tecnología siguió influyendo en desarrollos posteriores de la IA.

### **Avances en algoritmos y hardware**

Mientras los sistemas expertos dominaban las aplicaciones, los avances en **algoritmos** y **hardware** sentaron las bases de la IA moderna. Central en este progreso fue el resurgimiento de las **[redes neuronales artificiales](https://en.wikipedia.org/wiki/Artificial_neural_network)** (RNA), inspiradas en la estructura del cerebro humano.

- En 1986, el **[algoritmo de retropropagación](https://en.wikipedia.org/wiki/Backpropagation)** (backpropagation) —impulsado por investigadores como **[Geoffrey Hinton](https://en.wikipedia.org/wiki/Geoffrey_Hinton)**— revolucionó el entrenamiento de redes neuronales. La retropropagación permitía a las redes ajustar sus parámetros internos propagando el error hacia atrás a través de las capas, lo que les permitía aprender de sus errores. Esta innovación dio lugar a redes multicapa ***capaces de resolver problemas no lineales, como el reconocimiento de imágenes y el procesamiento del habla***.
- La época también vivió avances teóricos en **[deep learning](https://en.wikipedia.org/wiki/Deep_learning)**. En 1989, [Yann LeCun](https://scholar.google.com/citations?user=WLN3QrAAAAAJ&hl=es) presentó la arquitectura **[LeNet](https://en.wikipedia.org/wiki/LeNet)**, una red neuronal convolucional (CNN) capaz de [reconocer dígitos manuscritos](https://www.geeksforgeeks.org/machine-learning/handwritten-digit-recognition-using-neural-network/). Aunque limitados por la potencia de cálculo disponible, estos primeros modelos demostraron la viabilidad del deep learning. Investigadores como **[Jürgen Schmidhuber](https://en.wikipedia.org/wiki/Jürgen_Schmidhuber)** exploraron las **[redes neuronales recurrentes](https://en.wikipedia.org/wiki/Recurrent_neural_network)** (RNN), que procesaban datos secuenciales — un precursor de los modelos de lenguaje actuales.
- Los avances en hardware aceleraron aún más la IA. El auge del **[procesamiento paralelo](https://en.wikipedia.org/wiki/Parallel_computing)** y de chips especializados, como las **[GPU](https://en.wikipedia.org/wiki/Graphics_processing_unit)**, permitió cálculos más rápidos. En los 90, proyectos como la **[Connection Machine](https://en.wikipedia.org/wiki/Connection_Machine)** demostraron cómo el paralelismo masivo podía abordar tareas de IA. Estas innovaciones, aunque poco valoradas en su momento, prepararon el terreno para los avances del siglo XXI.

### **Hitos principales**

Dos logros históricos de los años 90 simbolizaron el creciente potencial de la IA: **el dominio del ajedrez** y **los vehículos autónomos**.

- En 1997, el **[Deep Blue](https://en.wikipedia.org/wiki/Deep_Blue_%28chess_computer%29)** de IBM venció al campeón mundial [Garry Kasparov](https://en.wikipedia.org/wiki/Garry_Kasparov) en un encuentro a seis partidas. `Deep Blue combinaba cálculo por fuerza bruta (evaluando 200 millones de posiciones por segundo) con heurísticas estratégicas`, demostrando la capacidad de la IA para superar a los humanos en dominios basados en reglas.
- Por su parte, proyectos como **[ALVINN](https://en.wikipedia.org/wiki/ALVINN)** (Autonomous Land Vehicle in a Neural Network) fueron pioneros en la conducción autónoma. Desarrollado en la Universidad Carnegie Mellon en 1989, ALVINN usaba una red neuronal para procesar las imágenes de una cámara y dirigir un vehículo. En 1995, el **[NavLab 5](https://en.wikipedia.org/wiki/Navlab)** de la CMU completó un viaje de 2.850 millas ("No Hands Across America"), conduciendo de forma autónoma el 98,2% de la distancia. Estos esfuerzos sentaron las bases de empresas como **[Waymo](https://en.wikipedia.org/wiki/Waymo)** y **[Tesla](https://en.wikipedia.org/wiki/Tesla,_Inc.)**.
- Otros hitos incluyeron **[TD-Gammon](https://en.wikipedia.org/wiki/TD-Gammon)** (1992), un programa de backgammon que aprendía por sí mismo, y **[Eigenfaces](https://en.wikipedia.org/wiki/Eigenface)** (1991), el primer enfoque estadístico de Turk y Pentland para el reconocimiento facial. Aunque estas tecnologías estaban en pañales, apuntaban al potencial futuro de la IA en juegos, robótica y biometría.

### **Legado de la época**

Los años 80 y 90 transformaron la IA de una curiosidad académica en una herramienta con impacto real. Aunque los sistemas expertos entraron en declive, popularizaron la IA en la industria e inspiraron los posteriores **[sistemas basados en el conocimiento](https://en.wikipedia.org/wiki/Knowledge-based_systems)**. Los avances en redes neuronales y hardware, aunque inicialmente marginales, se convirtieron en la base de innovaciones del siglo XXI como **[AlphaGo](https://en.wikipedia.org/wiki/AlphaGo)** y los modelos de lenguaje actuales. Las lecciones de esta época —equilibrar la ambición con la practicidad— siguen moldeando la evolución de la IA hoy en día.

## **El siglo XXI (2000–presente): la era del big data y el deep learning**

El siglo XXI ha vivido una explosión sin precedentes de [datos](https://en.wikipedia.org/wiki/Data) y [potencia de cálculo](https://en.wikipedia.org/wiki/Computational_power), impulsando los avances de la [inteligencia artificial (IA)](https://en.wikipedia.org/wiki/Artificial_intelligence). Con la proliferación de las [tecnologías digitales](https://en.wikipedia.org/wiki/Digital_technology), la disponibilidad de enormes [conjuntos de datos](https://en.wikipedia.org/wiki/Data_set) y el refinamiento de las técnicas de [machine learning](https://en.wikipedia.org/wiki/Machine_learning), la IA se ha convertido en parte integral de la vida moderna. Esta época ha estado definida en buena medida por dos componentes clave: el **[big data](https://en.wikipedia.org/wiki/Big_data)** y el **[deep learning](https://en.wikipedia.org/wiki/Deep_learning)**, que han moldeado de forma decisiva el desarrollo y la aplicación de la IA en múltiples ámbitos.

### **El papel del big data**

Una de las características que definen a la IA contemporánea es la enorme cantidad de [datos](https://en.wikipedia.org/wiki/Data) disponibles para entrenar modelos de machine learning. El auge de [internet](https://en.wikipedia.org/wiki/Internet), las [redes sociales](https://en.wikipedia.org/wiki/Social_media) y los [dispositivos conectados](https://en.wikipedia.org/wiki/Internet_of_things) ha provocado un aumento exponencial de la cantidad de datos estructurados y no estructurados. Esta explosión de datos ha sido clave para el avance de la IA por varias razones:

1. **[Mejora del entrenamiento de modelos](https://en.wikipedia.org/wiki/Training_data)**: la disponibilidad de grandes conjuntos de datos permite entrenar los sistemas de IA con ejemplos diversos, mejorando su precisión y su capacidad de generalización.
2. **[Toma de decisiones basada en datos](https://en.wikipedia.org/wiki/Data-informed_decision-making)**: los modelos de IA pueden analizar cantidades masivas de datos para extraer conclusiones, ayudando a empresas, gobiernos e investigadores a tomar decisiones informadas.
3. **[Experiencias personalizadas](https://en.wikipedia.org/wiki/Personalization)**: desde los sistemas de recomendación de las plataformas de streaming hasta la publicidad dirigida, el big data permite a la IA crear experiencias personalizadas para cada usuario.

#### **Aplicaciones del big data en la IA**

- **[Reconocimiento de imágenes](https://en.wikipedia.org/wiki/Image_recognition)**: conjuntos de datos a gran escala como [ImageNet](https://en.wikipedia.org/wiki/ImageNet) han impulsado avances en [visión por computador](https://en.wikipedia.org/wiki/Computer_vision), permitiendo aplicaciones como el [reconocimiento facial](https://en.wikipedia.org/wiki/Facial_recognition_system), los [vehículos autónomos](https://en.wikipedia.org/wiki/Self-driving_car) y la [imagen médica](https://en.wikipedia.org/wiki/Medical_imaging).
- **[Procesamiento del lenguaje natural (NLP)](https://en.wikipedia.org/wiki/Natural_language_processing)**: los sistemas de NLP basados en IA dependen de corpus de texto masivos para mejorar la comprensión del lenguaje, y son la base de aplicaciones como los [chatbots](https://en.wikipedia.org/wiki/Chatbot), los [asistentes virtuales](https://en.wikipedia.org/wiki/Virtual_assistant) y los [servicios de traducción automática](https://en.wikipedia.org/wiki/Machine_translation).
- **[Sistemas de recomendación](https://en.wikipedia.org/wiki/Recommender_system)**: la IA aprovecha el big data para analizar las preferencias de los usuarios y ofrecer recomendaciones personalizadas en el [comercio electrónico](https://en.wikipedia.org/wiki/E-commerce), el [entretenimiento](https://en.wikipedia.org/wiki/Streaming_media) y las [noticias](https://en.wikipedia.org/wiki/Online_news).

### **Deep learning: un enfoque revolucionario**

El [deep learning](https://en.wikipedia.org/wiki/Deep_learning), un subconjunto del [machine learning](https://en.wikipedia.org/wiki/Machine_learning), se ha convertido en el `enfoque dominante en la investigación y las aplicaciones de IA`. Consiste en entrenar [redes neuronales artificiales](https://en.wikipedia.org/wiki/Artificial_neural_network) con `múltiples capas (redes profundas)` para identificar patrones en los datos. A diferencia de las técnicas tradicionales de machine learning, `los modelos de deep learning extraen automáticamente las características de los datos en bruto, eliminando la necesidad de una extensa ingeniería manual de características`.

#### **Pioneros del deep learning**

Tres investigadores —[Geoffrey Hinton](https://en.wikipedia.org/wiki/Geoffrey_Hinton), [Yann LeCun](https://en.wikipedia.org/wiki/Yann_LeCun) y [Yoshua Bengio](https://en.wikipedia.org/wiki/Yoshua_Bengio)— están ampliamente considerados los pioneros del deep learning. Sus aportaciones han sido decisivas para moldear el campo:

- **[Geoffrey Hinton](https://en.wikipedia.org/wiki/Geoffrey_Hinton)**: desarrolló técnicas de [retropropagación](https://en.wikipedia.org/wiki/Backpropagation) para entrenar redes neuronales profundas y jugó un papel clave en el avance de las [deep belief networks](https://en.wikipedia.org/wiki/Deep_belief_network).
- **[Yann LeCun](https://en.wikipedia.org/wiki/Yann_LeCun)**: introdujo las [redes neuronales convolucionales (CNN)](https://en.wikipedia.org/wiki/Convolutional_neural_network), revolucionando el reconocimiento de imágenes y la [visión por computador](https://en.wikipedia.org/wiki/Computer_vision).
- **[Yoshua Bengio](https://en.wikipedia.org/wiki/Yoshua_Bengio)**: contribuyó de forma decisiva al [modelado de secuencias](https://en.wikipedia.org/wiki/Sequence_learning) y a las técnicas de optimización del deep learning.

#### **Avances en redes neuronales**

- **[Redes neuronales convolucionales (CNN)](https://en.wikipedia.org/wiki/Convolutional_neural_network)**: estas redes están especializadas en el análisis de imágenes y vídeo, detectando jerarquías espaciales en los datos. Las CNN han sido claves en el desarrollo de aplicaciones de IA en el [reconocimiento facial](https://en.wikipedia.org/wiki/Facial_recognition_system), el [diagnóstico médico](https://en.wikipedia.org/wiki/Medical_diagnosis) y los [vehículos autónomos](https://en.wikipedia.org/wiki/Self-driving_car).
- **[Redes neuronales recurrentes (RNN)](https://en.wikipedia.org/wiki/Recurrent_neural_network)**: a diferencia de las redes neuronales tradicionales, las RNN pueden procesar datos secuenciales, lo que las hace ideales para aplicaciones como el [reconocimiento de voz](https://en.wikipedia.org/wiki/Speech_recognition) y el [modelado del lenguaje](https://en.wikipedia.org/wiki/Language_model).

### **Conceptos básicos de la IA**

La inteligencia artificial (IA) es un campo de la informática centrado en desarrollar sistemas capaces de realizar tareas que normalmente requieren inteligencia humana, como el aprendizaje, el razonamiento, la resolución de problemas, la percepción y la toma de decisiones. Engloba varios subcampos y tecnologías, entre ellos el machine learning, el deep learning, el procesamiento del lenguaje natural, la visión por computador y la robótica, que trabajan juntos para `permitir que las máquinas aprendan de los datos, se adapten a nuevas situaciones y mejoren su rendimiento con el tiempo`.

1. **[Machine Learning (ML)](https://en.wikipedia.org/wiki/Machine_learning)**: un subconjunto central de la IA en el que los sistemas aprenden de los datos sin estar programados explícitamente. Implica algoritmos que identifican patrones en los datos para hacer predicciones o tomar decisiones. El ML se clasifica en aprendizaje supervisado (con datos etiquetados), aprendizaje no supervisado (encontrar patrones en datos sin etiquetar), aprendizaje semisupervisado (una mezcla de ambos) y aprendizaje por refuerzo (aprender por ensayo y error con recompensas y penalizaciones).
   Ejemplo: [predicción de fallos de equipos a partir de datos de sensores](https://spd.tech/machine-learning/predictive-maintenance/).

2. **[Deep Learning (DL)](https://en.wikipedia.org/wiki/Deep_learning)**: un subcampo del machine learning que usa redes neuronales artificiales con múltiples capas para procesar datos complejos como imágenes, audio y texto. Imita la estructura del cerebro humano y puede aprender automáticamente las características relevantes a partir de datos en bruto, mejorando su precisión con el tiempo.
   Ejemplo: [diseño topológico automatizado para mejorar la resistencia a daños de buques navales mediante deep learning geométrico](https://academic.oup.com/jcde/article/10/2/794/7084993?login=false).

3. **[Redes Neuronales Artificiales (RNA)](https://en.wikipedia.org/wiki/Artificial_neural_network)**: modelos computacionales inspirados en el cerebro humano, formados por nodos interconectados (neuronas) organizados en capas (entrada, ocultas, salida). Procesan la información ajustando pesos y umbrales, lo que permite al sistema aprender y hacer predicciones.
   Ejemplo: [sistemas de navegación autónoma para buques](https://promwad.com/news/navigation-systems-autonomous-ships).

4. **[Procesamiento del Lenguaje Natural (NLP)](https://en.wikipedia.org/wiki/Natural_language_processing)**: una rama de la IA que permite a las máquinas entender, interpretar y generar lenguaje humano. Es la base de aplicaciones como los asistentes de voz, los servicios de traducción y los chatbots, analizando la gramática, el contexto y el significado del texto y del habla.
   Ejemplo: [interfaces conversacionales para buques autónomos](https://www.sciencedirect.com/science/article/pii/S0029801824019796), que permiten a un Buque de Superficie Marítimo Autónomo (MASS) responder en lenguaje natural sobre su estado e intenciones.

5. **[IA generativa](https://en.wikipedia.org/wiki/Generative_artificial_intelligence)**: una aplicación de la IA que crea contenido nuevo y original, como texto, imágenes, audio, vídeo y código, identificando patrones en grandes conjuntos de datos. Se apoya en grandes modelos de lenguaje (LLM) y modelos fundacionales entrenados con enormes cantidades de datos, que les permiten generar resultados a partir de las instrucciones del usuario.
   Ejemplo: [la IA generativa en ingeniería naval: los conjuntos de datos pequeños y propietarios limitan su adopción](https://www.zdnet.com/article/generative-ai-in-naval-architecture-and-marine-engineering-small-proprietary-data-sets-limit-adoption/).

6. **[Agentes de IA](https://github.com/resources/articles/ai/what-are-ai-agents)**: entidades que perciben su entorno y actúan para lograr objetivos. Pueden clasificarse como asistidos (ayudan a los humanos), aumentados (potencian las capacidades humanas) o autónomos (operan de forma independiente sin intervención humana).
   Ejemplo: [revolucionando el diseño de buques con inteligencia artificial y machine learning](https://www.bmt.org/insights/revolutionising-ship-design-with-artificial-intelligence-and-machine-learning/).

### **Tipos de IA: IA débil frente a IA fuerte**

La Inteligencia Artificial (IA) se clasifica a grandes rasgos en dos tipos principales: [IA débil e IA fuerte](https://www.geeksforgeeks.org/artificial-intelligence/what-is-the-difference-between-a-strong-ai-and-a-weak-ai/), que representan capacidades y niveles de inteligencia distintos.

| Característica | **IA débil** | **IA fuerte** |
|----------------------|--------------------------------|------------------------------------|
| Alcance | Estrecho, específico de una tarea | Amplio, de propósito general |
| Consciencia | No | Teóricamente, sí |
| Pensamiento a nivel humano | No | Sí |
| Ejemplos | Siri, Google Translate, filtros de spam | HAL 9000, Data (*ficticios*) |
| Aprendizaje y adaptación | Limitado | Flexible y con mejora autónoma |

| Tipo | Descripción | Perspectiva marítima |
|------|-------------|----------------------|
| IA débil (Narrow AI) | Especializada en una tarea. No puede generalizar más allá de ella. | Sistema de piloto automático, asistente de navegación. |
| IA fuerte (IA general) | Inteligencia a nivel humano, flexible y adaptable. | Hipotética — un buque totalmente autónomo capaz de tomar decisiones como un capitán en cualquier escenario. |

> **Idea clave**: la mayoría de los sistemas de IA actuales — incluido todo lo que construiremos en este curso — son **IA débil**. La IA fuerte sigue siendo teórica y continúa siendo un reto de investigación abierto.

### **Ejemplo práctico: predicción de fallos por vibraciones del motor**

Vamos a simular un caso simplificado de mantenimiento predictivo: un sensor de vibración del motor de un buque, que registra datos cada hora durante un día.

> **Para saber más**: [Detección de anomalías (Wikipedia)](https://en.wikipedia.org/wiki/Anomaly_detection) · [Vibración (Wikipedia)](https://en.wikipedia.org/wiki/Vibration).

```python
# Let's Visualize a Use Case
import matplotlib.pyplot as plt
import numpy as np

# Simulated engine vibration data
np.random.seed(0)
hours = np.arange(0, 24, 1)
vibration = 0.3 * np.sin(0.3 * hours) + np.random.normal(0, 0.1, size=hours.shape)

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, marker='o')
plt.title("Simulated Engine Vibration Over 24 Hours")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.grid(True)
plt.show()
```

Si etiquetamos cuándo la vibración supera un umbral, la IA puede aprender a predecir fallos con antelación.

```python
thresh = 0.25
alerts = vibration > thresh

plt.figure(figsize=(10, 4))
plt.plot(hours, vibration, label="Vibration")
plt.axhline(thresh, color='red', linestyle='--', label="Threshold")
plt.scatter(hours[alerts], vibration[alerts], color='red', label="Alerts")
plt.title("Predictive Maintenance Threshold")
plt.xlabel("Time (hours)")
plt.ylabel("Vibration Amplitude")
plt.legend()
plt.grid(True)
plt.show()
```

**Pruébalo tú mismo**: cambia `thresh` a un valor más bajo (p. ej., `0.15`) y compara cuántas horas quedan marcadas. Un umbral demasiado bajo corre el riesgo de saturar a un equipo de mantenimiento real con falsas alarmas; uno demasiado alto corre el riesgo de pasar por alto un aviso temprano real — no hay un único valor "correcto", es un compromiso de ingeniería real con el que se topará de nuevo cada sistema de detección de anomalías de este curso.

```python
thresh_lower = 0.15
alerts_lower = vibration > thresh_lower

print(f"Alerts at threshold {thresh}: {alerts.sum()} hours")
print(f"Alerts at threshold {thresh_lower}: {alerts_lower.sum()} hours")
```
