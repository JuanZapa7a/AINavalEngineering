## Part 1 — Python Language Refresher *(≈25 min · Theory + Practice)*

Everything from here on assumes you can read and write basic Python. If you already know it well, treat this as a fast warm-up; if not, this is the minimum toolkit you need before the rest of the course.

### Variables and Basic Types

Python is dynamically typed: a variable's type is inferred from the value it's assigned, and can change if reassigned.

| Type | Example | Naval-flavored example |
|---|---|---|
| `int` | `n = 4` | `num_engines = 2` |
| `float` | `x = 3.14` | `length_m = 209.9` |
| `str` | `s = "hello"` | `ship_name = "MV Poseidon"` |
| `bool` | `b = True` | `is_operational = True` |


```python
ship_name = "MV Poseidon"
length_m = 209.9
num_engines = 2
is_operational = True

print(ship_name, length_m, num_engines, is_operational)
print(type(ship_name), type(length_m), type(num_engines), type(is_operational))
```

### Collections: Lists, Tuples, and Dictionaries

- A **list** (`[...]`) is an ordered, mutable sequence.
- A **tuple** (`(...)`) is like a list but immutable (can't be changed after creation).
- A **dict** (`{...}`) maps keys to values — ideal for representing one record (e.g., one ship) with named fields.


```python
fleet = ["MV Poseidon", "SS Neptune", "MV Triton", "SS Amphitrite"]
print(fleet[0])       # first element
print(fleet[-1])      # last element
print(fleet[1:3])     # slicing: elements 1 and 2
print(len(fleet))     # number of elements
```

Now a **dict**, to represent one ship as a single record with named fields:

```python
ship = {
    "name": "MV Poseidon",
    "length_m": 209.9,
    "displacement_t": 51252.5,
    "type": "Cargo Ship",
}

print(ship["name"], "-", ship["type"])
ship["is_operational"] = True  # dicts are mutable: we can add new keys
print(ship)
```

### Control Flow: `if`, `for`, `while`

Python uses **indentation** (not braces) to mark code blocks.


```python
for name, length in zip(fleet, [209.9, 320.0, 45.3, 180.2]):
    if length > 200:
        print(f"{name}: {length} m -> large vessel")
    elif length > 100:
        print(f"{name}: {length} m -> medium vessel")
    else:
        print(f"{name}: {length} m -> small vessel")
```

### Functions

Functions package reusable logic, take parameters, and (usually) `return` a value.


```python
def classify_by_displacement(displacement_t):
    """Classify a vessel by its displacement in tonnes."""
    if displacement_t > 50000:
        return "large"
    elif displacement_t > 10000:
        return "medium"
    else:
        return "small"

for d in [51252.5, 15000, 3000]:
    print(d, "->", classify_by_displacement(d))
```

### List Comprehensions

A compact, very common way to build a new list from an existing one, optionally filtering it.


```python
lengths = [209.9, 320.0, 45.3, 180.2, 260.5]

# Classic loop
large_ships = []
for length in lengths:
    if length > 150:
        large_ships.append(length)

# Equivalent list comprehension
large_ships_comp = [length for length in lengths if length > 150]

print(large_ships)
print(large_ships_comp)
print(large_ships == large_ships_comp)
```

### **The Python Interpreter**

The standard way to execute Python code is to run the program directly through the `Python interpreter`. On most systems, the Python interpreter is invoked using the `python` command. When a Python source file is passed as an argument to this command, the Python code in the file is executed.

In Google Colab, we can create, write, and execute a Python script directly from a notebook cell:

- **`%%writefile hello.py`**: a *magic command* that creates a new file named `hello.py` and writes the following lines into it.
- **`!python hello.py`**: the `!` symbol executes a **shell command** directly from the notebook — here, it runs `hello.py` with the Python interpreter.
- **`!python --version`**: checks the version of Python installed in the current environment.


```python
%%writefile hello.py
print("Hello from Python!")
```

Run the script we just wrote:

```python
!python hello.py
```

Check which Python version Colab is running:

```python
!python --version
```

### **Input and Output Caching**

In the IPython console (a Google Colab code cell, for us), the input prompt is denoted as `In [1]:` and the corresponding output as `Out [1]:`, where the numbers increment for each new input/output. Both are stored in the `In` (a list) and `Out` (a dictionary) variables, so previous inputs/outputs can be accessed by cell number. Output can be suppressed by ending a statement with a semicolon (`;`).

### **IPython Magic Commands**

IPython provides *magic functions*, all starting with one or two `%` signs: a single `%` for one-line commands, `%%` for commands that operate on a whole cell. `%lsmagic` lists them all; `<magic>?` shows the docs for one.

`%run` executes an external script file *within* the current interactive session — unlike `!python file.py`, it also adds the script's symbols (functions, variables) to your notebook's namespace, so you can keep exploring them afterwards.


```python
%lsmagic
```

Let's put `%run` to use with a small Fibonacci script:

```python
%%writefile fib.py

def fib(N):
    """
    Return a list of the first N Fibonacci numbers.
    """
    f0, f1 = 0, 1
    f = [1] * N
    for n in range(1, N):
        f[n] = f0 + f1
        f0, f1 = f1, f[n]

    return f

print(fib(10))
```

Run it with `%run` — unlike `!python`, this also loads `fib` into our notebook's namespace:

```python
%run fib.py
```

Because `%run` imported `fib`, we can call it directly, no import needed:

```python
fib(15)
```

**Exercise**: create a new version of the Fibonacci sequence that includes 0. Name it `fib2.py` and write it using `%%writefile`, exactly as above.


```python
%%writefile fib2.py

def fib(N):
    """
    Return a list of the first N Fibonacci numbers starting from 0.
    """
    if N <= 0:
        return []

    f0, f1 = 0, 1
    f = [0] * N
    if N > 1:
        f[1] = 1

    for n in range(2, N):
        f[n] = f0 + f1
        f0, f1 = f1, f[n]

    return f
```

Run the corrected version:

```python
%run fib2.py
fib(15)
```

```python
%who
```

## Debugging, Reset, and Timing/Profiling *(≈15 min · Practice)*

### **Debugger**

IPython includes a `postmortem` debugger: after an uncaught exception is raised, typing `%debug` drops you into the Python debugger *at the point of failure*, with access to the local namespace — no need to sprinkle `print()` statements or rerun a long computation from scratch.


```python
# This raises an error because fib expects an integer
fib(1.0)
```

That raised an exception — jump into the debugger at the exact point of failure:

```python
%debug
```

### **Reset**

`%reset -f` clears the IPython namespace (forcing it, without confirmation) — useful to make sure a script runs in a clean environment. Note: after `%reset`, re-importing an already-changed module reuses IPython's cached version, not necessarily the latest one on disk.

### **Timing and Profiling Code**

`%timeit` (and `%%timeit` for a whole cell) benchmarks a statement by running it multiple times and reporting the average runtime — handy for spotting bottlenecks. For a detailed breakdown of *where* time is spent, use the profiler: `%prun` for a single statement.

Example: simulate `N` random ships, each taking a random walk of `M` steps, and find the largest distance any of them reaches from the starting point.


```python
%reset -f
```

`fib` is gone after the reset — reload it, then benchmark it with `%timeit`:

```python
%run fib2.py  # reload fib after the reset
%timeit fib(100)
```

For a more realistic profiling example, define a small NumPy simulation: N random ships, each taking a random walk of M steps:

```python
import numpy as np

def random_ship_max_distance(M, N):
    """
    Simulate N random ships taking M miles, and return the largest distance
    from the starting point achieved by any of the random walkers.
    """
    trajectories = [np.random.randn(M).cumsum() for _ in range(N)]
    return np.max(np.abs(trajectories))
```

Profile it with `%prun` to see where the time actually goes:

```python
%prun random_ship_max_distance(400, 10000)
```

The profiler output breaks down how many times each function was called and where the cumulative time went — in this example, most of the time is spent inside `np.random.randn`.

## Part 3 — NumPy Basics for Data Work *(≈15 min · Theory + Practice)*

[NumPy](https://numpy.org/) is the numerical backbone under Pandas, scikit-learn, and every deep learning library we'll use later. Its core object is the `ndarray`: a fast, homogeneous, multi-dimensional array (a full deep dive into shapes, dtypes, and matrix operations is beyond today's scope; here we only cover what we need for the rest of the course).

### Creating Arrays and Basic Attributes


```python
import numpy as np

lengths_m = np.array([209.9, 197.2, 213.0, 230.5, 195.3])
print(lengths_m)
print("shape:", lengths_m.shape, "| ndim:", lengths_m.ndim, "| dtype:", lengths_m.dtype)
```

### Indexing, Slicing, and Boolean Masks

Arrays are indexed like Python lists, but also support **boolean masking**: indexing with a condition returns only the elements where the condition is `True` — a pattern you'll use constantly with real datasets.


```python
print(lengths_m[0])         # first element
print(lengths_m[1:3])       # slice: elements 1 and 2
print(lengths_m > 200)      # boolean mask
print(lengths_m[lengths_m > 200])  # only the "large" ships
```

### Vectorized Operations and Basic Statistics

NumPy operations apply **elementwise**, without writing an explicit loop — this is both more concise and much faster than pure Python (recall `%timeit` from Part 2).


```python
lengths_ft = lengths_m * 3.28084  # meters to feet, applied to every element at once
print(lengths_ft.round(1))

print("mean:", lengths_m.mean().round(1), "m")
print("std:", lengths_m.std().round(1), "m")
print("min/max:", lengths_m.min(), "/", lengths_m.max(), "m")
```

### 2D Arrays: Rows as Ships, Columns as Features

A 2D array is the NumPy equivalent of a small table — conceptually the same shape a Pandas DataFrame will use in Part 4, just without column names.


```python
# columns: [length_m, beam_m, draft_m]
fleet_specs = np.array([
    [209.9, 33.0, 9.9],
    [197.2, 30.8, 11.4],
    [213.0, 30.0, 11.5],
])

print(fleet_specs.shape)   # (3 ships, 3 features)
print(fleet_specs[:, 0])   # all lengths (first column)
print(fleet_specs.mean(axis=0))  # mean of each column (feature)
```

A Pandas `DataFrame` is essentially a 2D NumPy array with row and column *labels* — which is exactly what was missing from `fleet_specs` above.

### Loading a Real Naval Dataset

`Naval_Dataset.csv` contains 150 synthetic-but-realistic ship records: length, beam, draft, displacement, and ship type (Cargo Ship, Fishing Vessel, Passenger Ship). It has no header row, so we supply column names ourselves.


```python
!wget -q -O Naval_Dataset.csv https://raw.githubusercontent.com/JuanZapa7a/AINavalEngineering/main/Datasets/Naval_Dataset.csv

import pandas as pd

columns = ["Length_m", "Beam_m", "Draft_m", "Displacement_t", "Ship_Type"]
ships = pd.read_csv("Naval_Dataset.csv", names=columns)
ships.head()
```

### First-Pass Exploratory Data Analysis (EDA)

The same three questions apply to *every* dataset you'll meet in this course: **What columns do I have? What are their ranges? Are there missing values?**


```python
ships.info()
```

Now the numeric ranges:

```python
ships.describe()
```

And how many ships of each type:

```python
ships["Ship_Type"].value_counts()
```

### Filtering and Sorting

Boolean masking works on DataFrames exactly like it did on NumPy arrays above. `sort_values` orders rows by one or more columns.


```python
large_ships = ships[ships["Displacement_t"] > 55000]
print(f"{len(large_ships)} ships over 55,000 t")
large_ships.sort_values("Displacement_t", ascending=False).head()
```

### Grouping: How Do Ship Types Differ?

`groupby` lets us summarize numeric columns per category — a first, purely descriptive look at whether ship type correlates with size, before we ever train a model.


```python
ships.groupby("Ship_Type")[["Length_m", "Beam_m", "Draft_m", "Displacement_t"]].mean().round(1)
```

### Correlation Between Features

`.corr()` computes the pairwise correlation between numeric columns (from -1 to 1) — a quick way to spot which measurements move together, before deciding what to feed a model later in the course.


```python
ships[["Length_m", "Beam_m", "Draft_m", "Displacement_t"]].corr().round(2)
```

## Exploratory Plots *(≈10 min · Practice)*

Let's visualize length vs. displacement, colored by ship type, and the distribution of one variable.


```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 5))
for ship_type, group in ships.groupby("Ship_Type"):
    ax.scatter(group["Length_m"], group["Displacement_t"], label=ship_type, alpha=0.7)

ax.set_xlabel("Length (m)")
ax.set_ylabel("Displacement (t)")
ax.set_title("Ship Length vs. Displacement by Type")
ax.legend()
ax.grid(True)
plt.show()
```

And the distribution of a single variable:

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.hist(ships["Displacement_t"], bins=15, edgecolor="black")
ax.set_xlabel("Displacement (t)")
ax.set_ylabel("Number of ships")
ax.set_title("Distribution of Ship Displacement")
plt.show()
```

> **Takeaway**: load → inspect (`head`, `info`, `describe`) → filter/sort → group → correlate → visualize. You'll repeat this exact EDA workflow with every new dataset for the rest of the course, before any model is ever trained.
